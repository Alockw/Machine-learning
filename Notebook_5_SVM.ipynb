## <FONT SIZE=4 COLOR="purple"> 3.1 Matemáticas detrás del método </FONT>

El hiperplano óptimo tiene por ecuación:

$$\mathbf{w}^{T} \mathbf{x}+b=0$$

Donde $\mathbf{w}$ es un vector normal (ortogonal) al hiperplano y $b$ se conoce como el sesgo.

Si consideramos la función

$$f(x) = \mathbf{w}^{T} \mathbf{x}+b$$

La frontera de decisión es precisamente $f(x)=0$

Entonces, la predicción está dada por el signo de $f(x)$:

 - Si $f(x)>0$, se predice como **clase positiva** o simplemente 1.

 - Si $f(x)<0$, se predice como **clase negativa** o simplemente -1 (también puede ser cero 0)


Como deseamos obtener el margen más grande que separe las 2 clases vamos a considerar la distancia entre  hiperplanos que pasen en los puntos más cercanos de las dos clases (vectores de soporte).

Los vectores de soporte están en los márgenes

$$\mathbf{w}^{T}\mathbf{x}+b = 1 \quad  o \quad \mathbf{w}^{T}\mathbf{x}+b = -1$$

Luego, la zona entre los vectores de soporte es la banda del margen.

Los márgenes se definen como

1. Si $\mathbf{w}^{T}\mathbf{x}+b\geq
1$, la clase predicha $\hat{y}$ es la clase positiva (1).

2. Si $\mathbf{w}^{T}\mathbf{x}+b \leq -1$, la clase predicha $\hat{y}$ es la clase negativa (-1 o 0).

Ahora, el valor del margen se obtiene calculando la distancia de un punto a una de las rectas. La fórmula de esta distancia se puede calcular usando la distancia de un punto a un plano  (conceptos de álgebra lineal).

Recordemos que la distancia de un punto $x_0$ a un hiperplano es

$$d = \dfrac{|\mathbf{w}^{T}. x_0+b|}{\|w\|}$$

Además, para dos hiperplanos paralelos

$$\mathbf{w}^{T}\mathbf{x}+b_1 = 0 \qquad \mathbf{w}^{T}\mathbf{x}+b_2 =0$$

la distancia entre ellos es:

$$d = \dfrac{|b_1-b_2|}{\|w\|}$$

En este caso como $b_1=1$ y $b_2 =-1$

Entonces, la distancia entre ellos, que se denonima el margen es:


$$Margen=\frac{2}{\|\mathbf{w}\|}$$

Ahora bien, queremos que este margen sea lo más grande posible. Es decir, se convierte en un problema de optimización

$$\begin{align}
&\displaystyle \max_{\mathbf{w},b}\hspace{2em} Margen=\frac{2}{\|\mathbf{w}\|}\\
&\text{sujeto a}\hspace{1em} y_{i}\left(\mathbf{w}^{T}\mathbf{x}_{i}+b\right)\geq1\hspace{2em} \text{Para $i=1,2,\cdots,m$}
\end{align}$$

En cual lo podemos transformar en un problema de minimización

$$\begin{align}
&\displaystyle \min_{\mathbf{w},b}\hspace{2em} \frac{\|\mathbf{w}\|}{2}\\
&\text{sujeto a}\hspace{1em} y_{i}\left(\mathbf{w}^{T}\mathbf{x}_{i}+b\right)\geq1\hspace{2em} \text{Para $i=1,2,\cdots,m$}
\end{align}$$

Donde:

1. $y_{i}=-1$ para la clase negativa y $y_{i}=1$ para la clase positiva.

2. $\mathbf{x}_{i}$ es un registro del conjunto de datos y la igualdad de las restricciones se cumple para los vectores de soporte.



 ***Soft Margin Classification.***

Si imponemos estrictamente que todas las instancias estén fuera de la margen y del lado correcto, esto se denomina _hard margin classification_ (clasificación de margen duro). Hay dos problemas principales con el *hard margin classification*.

1. Solo funciona si los datos son linealmente separables.

2. Es bastante sensible a los valores atípicos.

Para evitar estos problemas, es preferible utilizar un modelo más flexible.

- El objetivo es encontrar un buen equilibrio entre mantener la margen lo más grande posible y limitar las violaciones de margen (es decir, instancias que terminan dentro de la margen o incluso incluso en el lado incorrecto). Esto se llama _soft margin classification_.

En los problemas de optimización se acostumbra regularizar. Esto es, penalizar el modelo para evitar que se ajuste demasiado a los datos (overfitting). Esta técnica modifica la función de costo al incluir un término que castiga los parámetros grandes o complejos

$$\begin{align}
&\displaystyle \min_{\mathbf{w},b,\mathbf{\xi}}\hspace{2em} \frac{\|w\|^2}{2}+C\displaystyle\sum_{i=1}^{m}\xi_i\\
&\text{sujeto a}\hspace{1em} y_{i}\left(\mathbf{w}^{T}\mathbf{x}_{i}+b\right)\geq1-\xi_i\hspace{0.4em} \text{y}\hspace{0.4em}\xi_i\geq0\hspace{1em} \text{Para $i=1,2,\cdots,m$}
\end{align}$$

En la clase [`SVM`](https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVC.html) de Scikit-Learn, se puede controlar este equilibrio usando el hiperparámetro $C$. De acuerdo con lo anterior se debe tener en cuenta

1. Un valor de $C$ más pequeño conduce a una ***margen*** más ancha pero conduce a más violaciones de la de margen.

2. Con un valor de $C$ alto el clasificador comete menos violaciones de ***margen*** pero la ***margen*** en sí será pequeña.

Note que en el caso de la situación 1, parece probable que ***generalice mejor*** a pesar que tiene "muchas" violaciones del margen.

***Sugerencia:***

Si un modelo de `SVM` presenta _overfitting_ se puede intentar regularizar reduciendo el valor de `C`.

$$C = \{0.001, 0.01, 0.1, 1 , 10, 100, 1000\}$$

# <FONT SIZE=4 COLOR="purple"> 4. Método SVM para clasificación (separación no lineal) Kernels</FONT>

¿Que podemos hacer cuando no existe la separación lineal?

Cuando no existe la separación lineal se puede usar alguna función matemática que permita transformar los datos, por ejemplo, aumentándole una dimensión (como el gráfico siguiente), de tal manera que se pueda clasificar. Estos tipos de funciones que transforman los datos se denominan ***kernel***

<center><img src="https://github.com/Fabian830348/cursos/blob/master/svm2.png?raw=true" alt="centered image" width="500" height="300"></center>





**Definición de Kernel**

Un kernel es una función definida como:

$$K(x_1,x_2):\mathbb{R}^{m}\times\mathbb{R}^{m}\rightarrow\mathbb{R},$$

con la particularidad de que puede ser expresado como un producto interno en otro espacio por medio de una función de mapeo

 $$ \varphi (x):\mathbb{R}^{m}\rightarrow\mathbb{R}^{k}$$

Es decir, se cumple la ecuación:

$$K(x_1,x_2)=\varphi(x_1)^{T}\varphi(x_2)$$

***Funciones kernel más usadas***

Para aumentar la dimensión de las variables y construir funciones más complejas se pueden usar:
1. Lineal
$$K(x_1,x_2)=x_1^{T}\cdot x_2$$

2. Polinómico (poly)
$$K(x_1,x_2)=(x_1^{T}\cdot x_2+a)^d$$
No se recomienda $d>5$ por _overffiting_.

3. Gaussiana (RBF)

$$K(x_1,x_2)=e^{-\gamma\|x_1-x_2\|^{2}}$$

4. Sigmoid

$$K(x_1,x_2)=\tanh\left(\gamma(x_1^{T}\cdot x_2)+r\right)$$

````python
# Dividir en dos conjuntos
# las variables predictoras
X = credito.drop("card", axis=1)
# la variable objetivo
y = credito["card"]
# conjunto de entrenamiento y de prueba
X_train, X_test, y_train, y_test = train_test_split(X,                        # variables predictoras
                                                    y,                        # variable de respuesta
                                                    stratify=y,               # estratificamos con respecto a y (asegura que se mantenga la proporción de clases en y)
                                                    random_state = 0,         # semilla para que al ejecutar siempre de igual
                                                    test_size = 0.3)          # tamaño del conjunto de prueba

# Definitmos el preprocesador de escalamiento
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), variables_num),
    ('cat', "passthrough", variables_cat) #passthrough
])
````

## <FONT SIZE=5 COLOR="blue"> 5.3 Generar los modelos de clasificación </FONT>

En esta sección usaremos los algoritmos:

- SVM
- k-vecinos más cercanos
- regresión logística
- árboles de decisión

````python
def evaluacion_modelo(modelo, X_train, y_train, X_test, y_test, nombre_modelo = "modelo",graficar = True):
  # definir las predicciones en conjuntos de train y test
  y_pred_train = modelo.predict(X_train)
  y_pred_test  = modelo.predict(X_test)
  # matrices de confusión
  if graficar :
    # Calcular matrices de confusión
    cm_train = confusion_matrix(y_train, y_pred_train)
    cm_test = confusion_matrix(y_test, y_pred_test)
    # ejes para gráficas
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    # matriz de confusión en train
    sns.heatmap(pd.DataFrame(cm_train),                # data.frame
                  annot=True,                          # colocar números de las cajitas
                  annot_kws = {'size':20},             # tamaño de la letra
                  cmap="YlOrRd",                       # color de la letra 'Pastel1', 'Pastel1_r', 'Pastel2', 'Pastel2_r', 'PiYG', 'PiYG_r', 'PuBu'
                  fmt='g',                             # para que salgan los número no : notación científica
                  ax=axes[0])
    axes[0].set_title(f'Matriz de Confusión - Train-{nombre_modelo}', fontsize=14)
    axes[0].set_xlabel('Predicciones')
    axes[0].set_ylabel('Valores Reales')
    # matriz de confusión en test
    sns.heatmap(pd.DataFrame(cm_test),
              annot=True,
              fmt='g',
              cmap='YlOrRd',
              annot_kws={'size':16},
              ax=axes[1])
    axes[1].set_title(f'Matriz de Confusión - Test-{nombre_modelo}', fontsize=14)
    axes[1].set_xlabel('Predicciones')
    axes[1].set_ylabel('Valores Reales')
    plt.tight_layout()
    plt.show()
  # metricas en train
  metrics=["accuracy", "recall" , "specificity", "precision", "f1"]
  # valores
  values_train = [accuracy_score(y_train,y_pred_train),
          recall_score(y_train,y_pred_train),
          specificity_score(y_train,y_pred_train),
          precision_score(y_train,y_pred_train),
          f1_score(y_train,y_pred_train)]
  values_test = [accuracy_score(y_test,y_pred_test),
          recall_score(y_test,y_pred_test),
          specificity_score(y_test,y_pred_test),
          precision_score(y_test,y_pred_test),
          f1_score(y_test,y_pred_test)]

  df = pd.DataFrame({f"{nombre_modelo}_train" : values_train,
                      f"{nombre_modelo}_test" : values_test}, index = metrics)
  return df
````

### <FONT SIZE=5 COLOR="magenta"> 5.3.1 SVM </FONT>

````python
# definimos el modelo de SVC
modelo_svc = Pipeline(steps=[
      ('preprocessor', preprocessor),
      ('classifier', SVC())
      ])
# entrenamos el modelo
modelo_svc.fit(X_train,y_train)
````

````python
evaluacion_modelo(modelo_svc, X_train, y_train, X_test, y_test, nombre_modelo= "SVM", graficar = True)
````

````python
param_grid = {
    'classifier__kernel': ['linear', 'rbf', 'poly'],
    'classifier__C': [0.1, 1, 10],
    #'classifier__gamma': ['scale', 'auto'],            # solo para 'rbf' y 'poly'
    #'classifier__degree': [2, 3, 4],                   # solo para 'poly'
}
grid_search = GridSearchCV(modelo_svc,
                           param_grid,
                           cv=5,
                           scoring='accuracy')
grid_search.fit(X_train, y_train)
````

````python
print("Mejores hiperparámetros encontrados:")
print(grid_search.best_params_)

print("\nMejor puntaje promedio (cross-validation):")
print(grid_search.best_score_)

# Modelo final entrenado
mejor_modelo = grid_search.best_estimator_
````

# <FONT SIZE=5 COLOR="purple"> 7. Observaciones sobre otros clasificadores </FONT>

1. Se puede usar la función ***LinearSVC()*** que es similar al SVC(kernel= "linear) . Sin embargo, tiene mayor flexibilidad para escoger parámetros como la norma con la cuál se hace la penalización: por ejemplo $l_1$ o $l_2$.

Se puede ver más documentación en:
[LinearSCV](https://scikit-learn.org/stable/modules/generated/sklearn.svm.LinearSVC.html)

2. Se puede usar también ***SGDClassifier()***.

   - Stochastic Gradient Descent - Classifier (SGD-Classifier)

   - El descenso de gradiente es uno de los algoritmos más populares para realizar la optimización y, con mucho, la forma más común de optimizar las redes neuronales. Pero también podemos usar este tipo de algoritmos para optimizar nuestro clasificador lineal, como la regresión logística y las máquinas de vectores de soporte lineal.

Hay tres tipos bien conocidos de descenso del gradiente

- **Descenso de gradiente por lotes**: calcula el gradiente utilizando todo el conjunto de datos para encontrar el mínimo ubicado en su cuenca de atracción.

- **Descenso de gradiente estocástico (SGD)**: calcula el gradiente usando una sola muestra.

- **Descenso de gradiente de mini lotes**:  realiza una actualización para cada minilote de n ejemplos de entrenamiento.
