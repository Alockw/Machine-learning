```python
import pulp as pl
import numpy as np
import matplotlib.pyplot as plt
import time
import pandas as pd
import seaborn as sns
from tabulate import tabulate
from matplotlib.ticker import MaxNLocator
import matplotlib.gridspec as gridspec
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
import matplotlib.dates as mdates
import matplotlib.ticker as mtick
from datetime import datetime, timedelta

# Configuración para gráficos más atractivos
plt.style.use('ggplot')
sns.set_palette("deep")
plt.rcParams['figure.figsize'] = [12, 8]
plt.rcParams['figure.dpi'] = 100
plt.rcParams['font.size'] = 10
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10
plt.rcParams['legend.fontsize'] = 10
plt.rcParams['figure.titlesize'] = 16

# Colores para cada recurso para mantener consistencia
COLORES = {
    "alimento": "#e41a1c",  # Rojo
    "madera": "#4daf4a",    # Verde
    "oro": "#ff7f00",       # Naranja
    "piedra": "#377eb8"     # Azul
}

# Iconos para recursos y unidades (emojis)
ICONOS = {
    "alimento": "🍗",
    "madera": "🌲",
    "oro": "💰",
    "piedra": "⛏️",
    "milicia": "⚔️",
    "arquero": "🏹",
    "lancero": "🔱",
    "caballero": "🐎",
    "mangudai": "🏇",
    "paladín": "🦄",
    "ciudadano": "👨‍🌾",
    "Oscura": "🌑",
    "Feudal": "🏠",
    "Castillos": "🏰",
    "Imperial": "👑",
}

def optimizar_aoe2(params=None):
    """
    Función principal para optimizar la estrategia en Age of Empires II

    Args:
        params: Diccionario con parámetros personalizados

    Returns:
        Resultados de la optimización
    """
    # Usar parámetros por defecto si no se proporcionan
    if params is None:
        params = {}

    # Parámetros del modelo con valores por defecto
    T_MAX = params.get('T_MAX', 15)  # Reducido a 15 para más eficiencia
    MAX_CIUDADANOS = params.get('MAX_CIUDADANOS', 30)
    INICIO_CIUDADANOS = params.get('INICIO_CIUDADANOS', 5)
    INICIO_RECURSOS = params.get('INICIO_RECURSOS',
                               {"alimento": 200, "madera": 200, "oro": 100, "piedra": 100})

    # Definiciones básicas
    EDADES = ["Oscura", "Feudal", "Castillos", "Imperial"]
    RECURSOS = ["alimento", "madera", "oro", "piedra"]

    # Unidades simplificadas para cada edad
    UNIDADES = {
        "Oscura": ["milicia"],
        "Feudal": ["arquero", "lancero"],
        "Castillos": ["caballero", "mangudai"],
        "Imperial": ["paladín"]
    }

    # Edificios requeridos por edad
    EDIFICIOS = {
        "Oscura": ["centro_urbano", "molino", "campamento_maderero", "campamento_minero"],
        "Feudal": ["arquería", "establo", "herrería", "mercado"],
        "Castillos": ["castillo", "monasterio", "taller_de_asedio"],
        "Imperial": ["universidad", "torre_de_bombardeo"]
    }

    # Costos de edificios
    COSTO_EDIFICIO = {
        "centro_urbano": {"madera": 275},
        "molino": {"madera": 100},
        "campamento_maderero": {"madera": 100},
        "campamento_minero": {"madera": 100},
        "arquería": {"madera": 175},
        "establo": {"madera": 175},
        "herrería": {"madera": 150},
        "mercado": {"madera": 175},
        "castillo": {"piedra": 650},
        "monasterio": {"madera": 175},
        "taller_de_asedio": {"madera": 200},
        "universidad": {"madera": 200, "piedra": 100},
        "torre_de_bombardeo": {"madera": 100, "piedra": 150}
    }

    # Tiempo de construcción (minutos)
    TIEMPO_CONSTRUCCION = {
        "centro_urbano": 2,
        "molino": 0.5,
        "campamento_maderero": 0.5,
        "campamento_minero": 0.5,
        "arquería": 0.5,
        "establo": 0.5,
        "herrería": 0.5,
        "mercado": 0.75,
        "castillo": 3,
        "monasterio": 0.5,
        "taller_de_asedio": 0.75,
        "universidad": 0.75,
        "torre_de_bombardeo": 1.5
    }

    # Tasas de recolección (por defecto o personalizadas)
    RECOLECCION = params.get('RECOLECCION', {
        "alimento": 25,
        "madera": 20,
        "oro": 22,
        "piedra": 18
    })

    # Costos de avance de edad
    COSTO_AVANCE = params.get('COSTO_AVANCE', {
        "Feudal": {"alimento": 500, "madera": 0, "oro": 0, "piedra": 0},
        "Castillos": {"alimento": 800, "madera": 200, "oro": 0, "piedra": 0},
        "Imperial": {"alimento": 1000, "madera": 800, "oro": 200, "piedra": 0}
    })

    # Tiempo para avanzar
    TIEMPO_AVANCE = params.get('TIEMPO_AVANCE', {
        "Feudal": 2,
        "Castillos": 3,
        "Imperial": 4
    })

    # Valor estratégico de cada edad
    VALOR_EDAD = params.get('VALOR_EDAD', {
        "Oscura": 1,
        "Feudal": 3,
        "Castillos": 6,
        "Imperial": 10
    })

    # Costos y utilidad de unidades (simplificados)
    COSTO_UNIDAD = params.get('COSTO_UNIDAD', {
        "milicia": {"alimento": 60, "madera": 0, "oro": 20, "piedra": 0},
        "arquero": {"alimento": 0, "madera": 25, "oro": 45, "piedra": 0},
        "lancero": {"alimento": 60, "madera": 25, "oro": 0, "piedra": 0},
        "caballero": {"alimento": 60, "madera": 0, "oro": 75, "piedra": 0},
        "mangudai": {"alimento": 0, "madera": 55, "oro": 65, "piedra": 0},
        "paladín": {"alimento": 60, "madera": 0, "oro": 75, "piedra": 0}
    })

    UTILIDAD_UNIDAD = params.get('UTILIDAD_UNIDAD', {
        "milicia": 1,
        "arquero": 1.5,
        "lancero": 1.2,
        "caballero": 3.5,
        "mangudai": 3,
        "paladín": 5
    })

    TIEMPO_ENTRENAMIENTO = params.get('TIEMPO_ENTRENAMIENTO', {
        "milicia": 0.3,
        "arquero": 0.35,
        "lancero": 0.3,
        "caballero": 0.4,
        "mangudai": 0.5,
        "paladín": 0.45
    })

    COSTO_CIUDADANO = params.get('COSTO_CIUDADANO', {"alimento": 50, "madera": 0, "oro": 0, "piedra": 0})

    # Preparar todas las unidades disponibles
    todas_unidades = []
    for edad in EDADES:
        todas_unidades.extend(UNIDADES[edad])
    todas_unidades = list(set(todas_unidades))

    # Crear problema de optimización
    prob = pl.LpProblem("Age_of_Empires_II_Optimization", pl.LpMaximize)

    # Rangos para iteración
    tiempo = range(1, T_MAX + 1)

    # ---- VARIABLES DE DECISIÓN ----
    print("Creando variables de decisión...")

    # Variables para ciudadanos
    ciudadanos = {}
    for r in RECURSOS:
        for t in tiempo:
            ciudadanos[r, t] = pl.LpVariable(f"ciudadanos_{r}_{t}", lowBound=0, cat='Integer')

    # Variables para recursos
    recursos = {}
    for r in RECURSOS:
        for t in range(T_MAX + 1):  # incluimos t=0 para recursos iniciales
            recursos[r, t] = pl.LpVariable(f"recursos_{r}_{t}", lowBound=0, cat='Continuous')

    # Variables para avance de edad
    avance = {}
    for e in EDADES[1:]:  # Excluimos la Edad Oscura ya que empezamos ahí
        for t in tiempo:
            avance[e, t] = pl.LpVariable(f"avance_{e}_{t}", cat='Binary')

    # Variables para indicar la edad actual
    en_edad = {}
    for e in EDADES:
        for t in range(T_MAX + 1):  # incluimos t=0 para estado inicial
            en_edad[e, t] = pl.LpVariable(f"en_edad_{e}_{t}", cat='Binary')

    # Variables para crear ciudadanos
    crear_ciudadano = {}
    for t in tiempo:
        crear_ciudadano[t] = pl.LpVariable(f"crear_ciudadano_{t}", lowBound=0, cat='Integer')

    # Variables para crear unidades militares (simplificadas)
    crear_unidad = {}
    for u in todas_unidades:
        for t in tiempo:
            crear_unidad[u, t] = pl.LpVariable(f"crear_unidad_{u}_{t}", lowBound=0, cat='Integer')

    # Variables para construir edificios
    construir_edificio = {}
    for e in EDIFICIOS:
        for b in EDIFICIOS[e]:
            for t in tiempo:
                construir_edificio[b, t] = pl.LpVariable(f"construir_{b}_{t}", cat='Binary')

    # ---- CONDICIONES INICIALES ----
    print("Estableciendo condiciones iniciales...")

    # En t=0, estamos en Edad Oscura
    prob += en_edad["Oscura", 0] == 1
    for e in EDADES[1:]:
        prob += en_edad[e, 0] == 0

    # Recursos iniciales
    for r in RECURSOS:
        prob += recursos[r, 0] == INICIO_RECURSOS[r]

    # Distribución inicial de ciudadanos
    prob += ciudadanos["alimento", 1] == INICIO_CIUDADANOS
    for r in ["madera", "oro", "piedra"]:
        prob += ciudadanos[r, 1] == 0

    # ---- FUNCIÓN OBJETIVO ----
    print("Definiendo la función objetivo...")

    # Maximizar valor estratégico (simplificado)
    objetivo = 0
    for e in EDADES:
        for t in tiempo:
            objetivo += VALOR_EDAD[e] * en_edad[e, t]

    for u in todas_unidades:
        for t in tiempo:
            objetivo += UTILIDAD_UNIDAD[u] * crear_unidad[u, t]

    # Valorar la construcción temprana de edificios
    for e in EDIFICIOS:
        for b in EDIFICIOS[e]:
            for t in tiempo:
                objetivo += (T_MAX - t) * 0.1 * construir_edificio[b, t]

    prob += objetivo

    # ---- RESTRICCIONES ----
    print("Estableciendo restricciones...")

    # 1. Balance de recursos
    for r in RECURSOS:
        for t in tiempo:
            gasto_avance = 0
            for e in EDADES[1:]:
                gasto_avance += COSTO_AVANCE[e][r] * avance[e, t]

            gasto_unidades = 0
            for u in todas_unidades:
                gasto_unidades += COSTO_UNIDAD[u][r] * crear_unidad[u, t]

            gasto_edificios = 0
            for e in EDIFICIOS:
                for b in EDIFICIOS[e]:
                    if r in COSTO_EDIFICIO[b]:
                        gasto_edificios += COSTO_EDIFICIO[b][r] * construir_edificio[b, t]

            gasto_ciudadanos = COSTO_CIUDADANO[r] * crear_ciudadano[t] if r == "alimento" else 0

            prob += recursos[r, t] == (recursos[r, t-1] +
                                     RECOLECCION[r] * ciudadanos[r, t] -
                                     gasto_avance -
                                     gasto_unidades -
                                     gasto_edificios -
                                     gasto_ciudadanos)

    # 2. Límite de ciudadanos
    for t in tiempo:
        suma_ciudadanos = 0
        for r in RECURSOS:
            suma_ciudadanos += ciudadanos[r, t]
        prob += suma_ciudadanos <= MAX_CIUDADANOS

    # 3. Requisitos de avance de edad
    for e in EDADES[1:]:
        for t in tiempo:
            for r in RECURSOS:
                if COSTO_AVANCE[e][r] > 0:
                    prob += recursos[r, t-1] >= COSTO_AVANCE[e][r] * avance[e, t]

            # Requisitos de edificios para avanzar
            edad_anterior = EDADES[EDADES.index(e) - 1]
            edificios_necesarios = 2  # Simplificación: necesitamos 2 edificios de la edad anterior

            suma_edificios = 0
            for b in EDIFICIOS[edad_anterior]:
                for tau in range(1, t):
                    suma_edificios += construir_edificio[b, tau]

            prob += suma_edificios >= edificios_necesarios * avance[e, t]

    # 4. Transición entre edades
    for e in EDADES[1:]:
        edad_anterior = EDADES[EDADES.index(e) - 1]

        for t in tiempo:
            # Solo se puede avanzar si estamos en la edad anterior
            prob += avance[e, t] <= en_edad[edad_anterior, t-1]

            # Efecto del avance después del tiempo necesario
            if t + TIEMPO_AVANCE[e] <= T_MAX:
                prob += en_edad[e, t + TIEMPO_AVANCE[e]] >= avance[e, t]

    # 5. En cada momento solo se puede estar en una edad
    for t in range(T_MAX + 1):
        suma_edades = 0
        for e in EDADES:
            suma_edades += en_edad[e, t]
        prob += suma_edades == 1

    # 6. Restricciones para construcción de unidades según la edad
    for u in todas_unidades:
        # Determinar en qué edades está disponible la unidad
        edades_disponibles = [e for e in EDADES if u in UNIDADES[e]]

        for t in tiempo:
            # Solo se pueden crear unidades de la edad actual o anteriores
            suma_edades_disponibles = 0
            for e in edades_disponibles:
                suma_edades_disponibles += en_edad[e, t-1]
            prob += crear_unidad[u, t] <= 10 * suma_edades_disponibles

            # Verificar recursos para crear la unidad
            for r in RECURSOS:
                if COSTO_UNIDAD[u][r] > 0:
                    prob += crear_unidad[u, t] * COSTO_UNIDAD[u][r] <= recursos[r, t-1]

    # 7. Balance de población de ciudadanos
    for t in range(2, T_MAX + 1):
        suma_ciudadanos_anterior = 0
        suma_ciudadanos_actual = 0
        for r in RECURSOS:
            suma_ciudadanos_anterior += ciudadanos[r, t-1]
            suma_ciudadanos_actual += ciudadanos[r, t]
        prob += suma_ciudadanos_actual == suma_ciudadanos_anterior + crear_ciudadano[t-1]

    # 8. Restricciones para construcción de edificios según la edad
    for t in tiempo:
        for e in EDIFICIOS:
            for b in EDIFICIOS[e]:
                # Solo se pueden construir edificios de la edad actual o anteriores
                suma_edades_disponibles = 0
                for e_actual in EDADES:
                    if EDADES.index(e_actual) >= EDADES.index(e):
                        continue
                    suma_edades_disponibles += en_edad[e_actual, t-1]
                prob += construir_edificio[b, t] <= suma_edades_disponibles

                # Verificar recursos para construir el edificio
                for r in COSTO_EDIFICIO[b]:
                    prob += construir_edificio[b, t] * COSTO_EDIFICIO[b][r] <= recursos[r, t-1]

    # ---- RESOLVER EL PROBLEMA ----
    print("Resolviendo el problema de optimización...")
    start_time = time.time()

    # Usar el solver con límite de tiempo para evitar largos cálculos
    solver = pl.PULP_CBC_CMD(msg=False, timeLimit=60)
    prob.solve(solver)

    elapsed_time = time.time() - start_time
    print(f"Tiempo de resolución: {elapsed_time:.2f} segundos")
    print(f"Estado: {pl.LpStatus[prob.status]}")

    if pl.LpStatus[prob.status] != "Optimal":
        print("No se encontró una solución óptima en el tiempo límite.")
        return None

    print(f"Valor óptimo: {pl.value(prob.objective)}")

    # Recopilar resultados
    resultados = {
        "status": pl.LpStatus[prob.status],
        "objetivo": pl.value(prob.objective),
        "recursos": {},
        "ciudadanos": {},
        "edades": [],
        "unidades": {},
        "edificios": {},
        "momentos_avance": {}
    }

    # Extraer valores de las variables
    for r in RECURSOS:
        resultados["recursos"][r] = [pl.value(recursos[r, t]) for t in range(T_MAX + 1)]
        resultados["ciudadanos"][r] = [pl.value(ciudadanos[r, t]) for t in tiempo]

    for t in range(T_MAX + 1):
        for e in EDADES:
            if pl.value(en_edad[e, t]) > 0.5:
                resultados["edades"].append(e)
                break

    for u in todas_unidades:
        unidades_por_tiempo = []
        for t in tiempo:
            valor = pl.value(crear_unidad[u, t])
            if valor > 0.5:
                unidades_por_tiempo.append((t, round(valor)))
        if unidades_por_tiempo:
            resultados["unidades"][u] = unidades_por_tiempo

    # Recopilar información de edificios
    for e in EDIFICIOS:
        for b in EDIFICIOS[e]:
            for t in tiempo:
                if pl.value(construir_edificio[b, t]) > 0.5:
                    if b not in resultados["edificios"]:
                        resultados["edificios"][b] = []
                    resultados["edificios"][b].append(t)

    # Recopilar información de avance de edad
    for e in EDADES[1:]:
        for t in tiempo:
            if pl.value(avance[e, t]) > 0.5:
                resultados["momentos_avance"][e] = t
                break

    # Añadir metadatos sobre la civilización
    resultados["parametros"] = params

    return resultados

def generar_guia_detallada(resultados):
    """
    Genera una guía detallada paso a paso basada en los resultados de la optimización

    Args:
        resultados: Diccionario con los resultados de optimización

    Returns:
        Lista de diccionarios con instrucciones detalladas por minuto
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return []

    edades = resultados.get("edades", [])
    momentos_avance = resultados.get("momentos_avance", {})
    ciudadanos = resultados.get("ciudadanos", {})
    recursos = resultados.get("recursos", {})
    edificios = resultados.get("edificios", {})
    unidades = resultados.get("unidades", {})

    # Crear una guía más detallada
    guia = []
    tiempo_max = len(edades)

    # Mapeo de edificios por edad para recomendaciones
    edificios_por_edad = {
        "Oscura": ["centro_urbano", "molino", "campamento_maderero", "campamento_minero"],
        "Feudal": ["arquería", "establo", "herrería", "mercado"],
        "Castillos": ["castillo", "monasterio", "taller_de_asedio"],
        "Imperial": ["universidad", "torre_de_bombardeo"]
    }

    # Crear acciones para cada minuto
    for t in range(tiempo_max):
        minuto = t
        acciones = []
        prioridad = []
        consejos = []

        # Verificar si existe el índice actual
        if t < len(edades):
            edad_actual = edades[t]

            # Cambio de edad
            if t > 0 and t - 1 < len(edades) and edades[t] != edades[t-1]:
                acciones.append(f"🎉 ¡Has alcanzado la Edad {edad_actual}!")

                # Añadir consejos específicos para la nueva edad
                if edad_actual == "Feudal":
                    consejos.append("Construye una arquería para producir arqueros y aumentar la presión militar temprana.")
                    consejos.append("Expande tu economía con un segundo centro urbano.")
                elif edad_actual == "Castillos":
                    consejos.append("Prioriza la construcción de un castillo para proteger recursos clave.")
                    consejos.append("Comienza a entrenar unidades de élite para formar un ejército poderoso.")
                elif edad_actual == "Imperial":
                    consejos.append("Mejora tus unidades a tecnologías imperiales para maximizar su potencial.")
                    consejos.append("Construye torres de bombardeo para reforzar tu defensa.")

            # Añadir inicio de investigación de avance de edad
            for e, momento in momentos_avance.items():
                if momento == t:
                    acciones.append(f"🔍 Iniciar investigación para avanzar a Edad {e}")

                    # Añadir recomendaciones de preparación para la siguiente edad
                    if e == "Feudal":
                        prioridad.append("Asegúrate de tener al menos 6-8 aldeanos en alimento para mantener producción.")
                    elif e == "Castillos":
                        prioridad.append("Prepara recursos para construir un castillo apenas llegues a la Edad de Castillos.")
                    elif e == "Imperial":
                        prioridad.append("Prepárate para mejorar tus unidades principales a tecnologías imperiales.")

            # Verificar cambios en ciudadanos
            if t > 0:
                for recurso in ciudadanos:
                    if t < len(ciudadanos[recurso]):
                        actual = int(ciudadanos[recurso][t])
                        anterior = int(ciudadanos[recurso][t-1]) if t-1 < len(ciudadanos[recurso]) else 0
                        diferencia = actual - anterior
                        if diferencia != 0:
                            icono = ICONOS.get(recurso, "")
                            accion = f"{icono} {'Añadir' if diferencia > 0 else 'Quitar'} {abs(diferencia)} aldeanos {'a' if diferencia > 0 else 'de'} {recurso}"
                            acciones.append(accion)

            # Verificar edificios construidos
            for edificio, momentos in edificios.items():
                if t in momentos:
                    nombre_edificio = edificio.replace('_', ' ').title()
                    acciones.append(f"🏗️ Construir {nombre_edificio}")

            # Verificar creación de unidades
            for unidad, momentos_cantidades in unidades.items():
                for momento, cantidad in momentos_cantidades:
                    if momento == t:
                        icono = ICONOS.get(unidad, "⚔️")
                        acciones.append(f"{icono} Entrenar {cantidad} {unidad.title()}")

            # Añadir consejos basados en el nivel de recursos
            for recurso, valores in recursos.items():
                if t < len(valores) and valores[t] > 400:
                    if recurso == "alimento" and edad_actual in ["Oscura", "Feudal"]:
                        consejos.append(f"Tienes exceso de alimento. Considera avanzar de edad o producir más unidades.")
                    elif recurso == "madera" and t > 5:
                        consejos.append(f"Tienes exceso de madera. Construye más edificios o traslada aldeanos a otros recursos.")
                    elif recurso == "oro" and edad_actual in ["Castillos", "Imperial"]:
                        consejos.append(f"Tienes oro acumulado. Aumenta la producción de unidades que requieren oro.")

            # Añadir consejos específicos basados en el minuto de juego
            if t == 0:
                consejos.append("Construye 2 casas inmediatamente y entrena aldeanos constantemente.")
            elif t == 3:
                consejos.append("Envía 4 aldeanos a recolectar bayas o a crear granjas.")
            elif t == 5:
                consejos.append("Envía 2-3 aldeanos a madera para mantener la producción de edificios.")
            elif t == 7 and edad_actual == "Oscura":
                consejos.append("Deberías estar cerca de avanzar a la Edad Feudal. Prepara recursos para construir inmediatamente.")
            elif t == 10 and edad_actual == "Feudal":
                consejos.append("Comienza a construir un ejército pequeño para explorar o defender.")

        # Añadir entrada solo si hay acciones válidas
        if acciones or prioridad or consejos:
            guia.append({
                "minuto": minuto + 1,  # Mostrar minutos empezando en 1 en lugar de 0
                "edad": edad_actual if t < len(edades) else None,
                "acciones": acciones,
                "prioridad": prioridad,
                "consejos": consejos
            })

    return guia

def crear_guia_por_edad(resultados):
    """
    Crea una guía estratégica organizada por edades

    Args:
        resultados: Diccionario con los resultados

    Returns:
        Diccionario con instrucciones por edad
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return {}

    # Extraer información de manera segura
    edades = resultados.get("edades", [])
    momentos_avance = resultados.get("momentos_avance", {})
    ciudadanos = resultados.get("ciudadanos", {})
    recursos = resultados.get("recursos", {})
    unidades = resultados.get("unidades", {})
    edificios = resultados.get("edificios", {})

    # Validar que tenemos datos mínimos
    if not edades or not ciudadanos:
        return {}

    # Crear estructura para guía por edad
    guia_por_edad = {}
    intervalos_edad = {}

    # Determinar los intervalos de cada edad
    edad_actual = edades[0]
    inicio_edad = 0

    for t, edad in enumerate(edades):
        if edad != edad_actual:
            intervalos_edad[edad_actual] = (inicio_edad, t - 1)
            edad_actual = edad
            inicio_edad = t

    # Añadir el último intervalo
    intervalos_edad[edad_actual] = (inicio_edad, len(edades) - 1)

    # Generar recomendaciones para cada edad
    for edad, (inicio, fin) in intervalos_edad.items():
        # Inicializar estructura para esta edad
        guia_por_edad[edad] = {
            "tiempo": f"Minutos {inicio+1}-{fin+1}",
            "distribucion_ciudadanos": {},
            "edificios_recomendados": [],
            "unidades_recomendadas": [],
            "objetivos_clave": [],
            "transicion": {}
        }

        # Distribución promedio de ciudadanos
        for recurso in ciudadanos:
            if inicio < len(ciudadanos[recurso]) and fin < len(ciudadanos[recurso]):
                valores = [int(ciudadanos[recurso][t]) for t in range(inicio, fin + 1)]
                if valores:
                    promedio = sum(valores) / len(valores)
                    guia_por_edad[edad]["distribucion_ciudadanos"][recurso] = round(promedio)

        # Edificios recomendados
        for edificio, momentos in edificios.items():
            for t in momentos:
                if inicio <= t <= fin:
                    nombre_edificio = edificio.replace('_', ' ').title()
                    if nombre_edificio not in guia_por_edad[edad]["edificios_recomendados"]:
                        guia_por_edad[edad]["edificios_recomendados"].append(nombre_edificio)

        # Unidades recomendadas
        for unidad, momentos_cantidades in unidades.items():
            for momento, cantidad in momentos_cantidades:
                if inicio <= momento <= fin:
                    if unidad not in [u["tipo"] for u in guia_por_edad[edad]["unidades_recomendadas"]]:
                        guia_por_edad[edad]["unidades_recomendadas"].append({
                            "tipo": unidad,
                            "cantidad": cantidad
                        })

        # Objetivos clave según la edad
        if edad == "Oscura":
            guia_por_edad[edad]["objetivos_clave"] = [
                "Construir una economía sólida con al menos 15-20 aldeanos",
                "Explorar el mapa para localizar recursos",
                "Avanzar a la Edad Feudal lo antes posible"
            ]
        elif edad == "Feudal":
            guia_por_edad[edad]["objetivos_clave"] = [
                "Construir edificios militares clave",
                "Comenzar a entrenar unidades básicas para defensa",
                "Expandirse con un segundo centro urbano",
                "Prepararse para avanzar a la Edad de Castillos"
            ]
        elif edad == "Castillos":
            guia_por_edad[edad]["objetivos_clave"] = [
                "Construir un castillo para proteger recursos clave",
                "Aumentar la presión militar con unidades de élite",
                "Expandir la economía con más centros urbanos",
                "Explorar las tecnologías únicas de tu civilización"
            ]
        elif edad == "Imperial":
            guia_por_edad[edad]["objetivos_clave"] = [
                "Maximizar la producción militar con todas las mejoras",
                "Asegurar posiciones estratégicas con fortificaciones",
                "Mantener la presión económica y militar",
                "Prepararse para el juego tardío y batallas decisivas"
            ]

        # Información sobre transición a la siguiente edad
        if edad != "Imperial":
            proxima_edad = [e for e in ["Feudal", "Castillos", "Imperial"] if e > edad][0]
            if proxima_edad in momentos_avance:
                momento_avance = momentos_avance[proxima_edad]
                guia_por_edad[edad]["transicion"] = {
                    "edad_objetivo": proxima_edad,
                    "tiempo_recomendado": momento_avance,
                    "recomendaciones": []
                }

                # Añadir recomendaciones específicas
                if proxima_edad == "Feudal":
                    guia_por_edad[edad]["transicion"]["recomendaciones"] = [
                        "Tener al menos 8-10 aldeanos en alimento",
                        "Construir 2 edificios de la Edad Oscura",
                        "Preparar recursos para construir inmediatamente en Feudal"
                    ]
                elif proxima_edad == "Castillos":
                    guia_por_edad[edad]["transicion"]["recomendaciones"] = [
                        "Tener al menos 14-16 aldeanos en recursos variados",
                        "Construir 2 edificios de la Edad Feudal",
                        "Ahorrar piedra para un castillo inmediato"
                    ]
                elif proxima_edad == "Imperial":
                    guia_por_edad[edad]["transicion"]["recomendaciones"] = [
                        "Tener una economía sólida con 30+ aldeanos",
                        "Construir 2 edificios de la Edad de Castillos",
                        "Preparar recursos para mejoras imperiales"
                    ]

    return guia_por_edad

def visualizar_resultados(resultados, civilizacion=None):
    """
    Crea visualizaciones mejoradas para los resultados de la optimización

    Args:
        resultados: Diccionario con los resultados
        civilizacion: Nombre de la civilización (opcional)
    """
    if resultados is None or resultados.get("status") != "Optimal":
        print("No hay resultados óptimos para visualizar.")
        return

    # Extraer datos
    recursos = resultados.get("recursos", {})
    ciudadanos = resultados.get("ciudadanos", {})
    edades = resultados.get("edades", [])
    unidades = resultados.get("unidades", {})

    # Configurar el título
    titulo_base = "Optimización de Estrategia AoE II"
    if civilizacion:
        titulo_base += f" - {civilizacion}"

    # Crear figura con subplots
    fig = plt.figure(figsize=(16, 12))
    gs = gridspec.GridSpec(3, 2, height_ratios=[2, 1, 1])

    # 1. Gráfico de evolución de recursos
    ax1 = plt.subplot(gs[0, 0])
    for recurso, valores in recursos.items():
        ax1.plot(range(len(valores)), valores, color=COLORES[recurso], linewidth=2.5,
                 label=f"{ICONOS.get(recurso, '')} {recurso.title()}")

    # Añadir líneas verticales para cambios de edad
    edades_unicas = []
    for i, edad in enumerate(edades):
        if i == 0 or edad != edades[i-1]:
            edades_unicas.append((i, edad))
            if i > 0:
                ax1.axvline(x=i, color='gray', linestyle='--', alpha=0.7)
                ax1.text(i, max([max(vals) for vals in recursos.values()]) * 0.9,
                         f"Edad {edad}", rotation=90, alpha=0.7)

    ax1.set_title("Evolución de Recursos", fontsize=14, fontweight='bold')
    ax1.set_xlabel("Minuto de juego", fontsize=12)
    ax1.set_ylabel("Cantidad", fontsize=12)
    ax1.legend(loc='upper left')
    ax1.grid(True, alpha=0.3)
    ax1.set_xlim(-0.5, len(edades) - 0.5)
    ax1.xaxis.set_major_locator(MaxNLocator(integer=True))

    # 2. Gráfico de distribución de ciudadanos
    ax2 = plt.subplot(gs[0, 1])

    # Apilar la distribución de ciudadanos
    datos_ciudadanos = {}
    for recurso, valores in ciudadanos.items():
        datos_ciudadanos[recurso] = valores

    df = pd.DataFrame(datos_ciudadanos)
    df.plot.area(ax=ax2, stacked=True, color=[COLORES[r] for r in df.columns], alpha=0.7)

    # Añadir líneas verticales para cambios de edad
    for i, edad in edades_unicas:
        if i > 0:
            ax2.axvline(x=i, color='gray', linestyle='--', alpha=0.7)
            ax2.text(i, df.sum(axis=1).max() * 0.9, f"Edad {edad}", rotation=90, alpha=0.7)

    ax2.set_title("Distribución de Ciudadanos", fontsize=14, fontweight='bold')
    ax2.set_xlabel("Minuto de juego", fontsize=12)
    ax2.set_ylabel("Número de ciudadanos", fontsize=12)
    ax2.legend(loc='upper left')
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim(-0.5, len(edades) - 0.5)
    ax2.xaxis.set_major_locator(MaxNLocator(integer=True))

    # 3. Gráfico de unidades producidas
    ax3 = plt.subplot(gs[1, :])

    # Crear matriz para conteo de unidades por tiempo
    max_tiempo = len(edades)
    todas_unidades = list(unidades.keys())

    if todas_unidades:
        matriz_unidades = np.zeros((len(todas_unidades), max_tiempo))

        for i, unidad in enumerate(todas_unidades):
            for t, cantidad in unidades.get(unidad, []):
                if t < max_tiempo:
                    matriz_unidades[i, t] = cantidad

        # Crear gráfico de barras apiladas
        bottom = np.zeros(max_tiempo)
        for i, unidad in enumerate(todas_unidades):
            ax3.bar(range(max_tiempo), matriz_unidades[i], bottom=bottom,
                   label=f"{ICONOS.get(unidad, '⚔️')} {unidad.title()}", alpha=0.7)
            bottom += matriz_unidades[i]

        # Añadir líneas verticales para cambios de edad
        for i, edad in edades_unicas:
            if i > 0:
                ax3.axvline(x=i, color='gray', linestyle='--', alpha=0.7)

        ax3.set_title("Producción de Unidades", fontsize=14, fontweight='bold')
        ax3.set_xlabel("Minuto de juego", fontsize=12)
        ax3.set_ylabel("Cantidad", fontsize=12)
        ax3.legend(loc='upper left', ncol=3)
        ax3.grid(True, alpha=0.3)
        ax3.set_xlim(-0.5, max_tiempo - 0.5)
        ax3.xaxis.set_major_locator(MaxNLocator(integer=True))
    else:
        ax3.text(0.5, 0.5, "No hay datos de unidades disponibles",
                horizontalalignment='center', verticalalignment='center',
                transform=ax3.transAxes, fontsize=12)

    # 4. Línea de tiempo de edades y eventos clave
    ax4 = plt.subplot(gs[2, :])

    # Definir puntos de cambio de edad y eventos clave
    eventos = []
    for e, t in resultados.get("momentos_avance", {}).items():
        eventos.append((t, f"Inicia avance a {e}"))
        # Calcular aproximadamente cuándo llegará a la edad siguiente
        tiempo_avance = {"Feudal": 2, "Castillos": 3, "Imperial": 4}
        eventos.append((t + tiempo_avance.get(e, 0), f"Llega a {e}"))

    # Ordenar eventos por tiempo
    eventos.sort(key=lambda x: x[0])

    # Crear línea de tiempo
    if eventos:
        for i, (tiempo, evento) in enumerate(eventos):
            if tiempo < max_tiempo:
                ax4.annotate(evento, xy=(tiempo, 0), xytext=(tiempo, 0.2 + (i % 3) * 0.2),
                           arrowprops=dict(arrowstyle='->', color='blue', alpha=0.6),
                           ha='center', fontsize=9)

    # Añadir barras de colores para las edades
    colores_edades = {"Oscura": "#555555", "Feudal": "#8B4513", "Castillos": "#4169E1", "Imperial": "#FFD700"}
    edad_actual = None
    inicio = 0

    for i, edad in enumerate(edades):
        if edad != edad_actual:
            if edad_actual:
                ax4.barh(y=0, width=i-inicio, left=inicio, height=0.2,
                       color=colores_edades.get(edad_actual, "#999999"),
                       alpha=0.6, label=f"Edad {edad_actual}" if i == len(edades)-1 or edad_actual != edades[i+1] else "")
            edad_actual = edad
            inicio = i

    # Añadir la última edad
    if edad_actual:
        ax4.barh(y=0, width=len(edades)-inicio, left=inicio, height=0.2,
               color=colores_edades.get(edad_actual, "#999999"),
               alpha=0.6, label=f"Edad {edad_actual}")

    ax4.set_title("Línea de Tiempo de Edades y Eventos Clave", fontsize=14, fontweight='bold')
    ax4.set_xlabel("Minuto de juego", fontsize=12)
    ax4.set_yticks([])
    ax4.set_xlim(-0.5, max_tiempo - 0.5)
    ax4.xaxis.set_major_locator(MaxNLocator(integer=True))

    # Crear leyenda para edades
    handles = [Patch(color=colores_edades[e], alpha=0.6, label=f"Edad {e}") for e in colores_edades if e in edades]
    ax4.legend(handles=handles, loc='upper right', ncol=4)

    # Ajustar layout
    plt.suptitle(titulo_base, fontsize=18, fontweight='bold')
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)

    plt.show()

    # Crear un gráfico adicional para la estrategia económica
    plt.figure(figsize=(14, 8))

    # Crear una matriz de datos para el heatmap
    recursos_data = {}
    for r in recursos:
        if len(recursos[r]) > 0:
            recursos_data[r] = recursos[r]

    df_recursos = pd.DataFrame(recursos_data)

    # Normalizar los datos para el heatmap
    df_norm = (df_recursos - df_recursos.min()) / (df_recursos.max() - df_recursos.min())

    # Crear el heatmap
    ax = sns.heatmap(df_norm.T, cmap="YlOrRd", cbar_kws={'label': 'Nivel de recurso (normalizado)'})

    # Configurar el gráfico
    plt.title("Estrategia Económica - Evolución de Recursos", fontsize=16, fontweight='bold')
    plt.xlabel("Minuto de juego", fontsize=12)
    plt.ylabel("Recurso", fontsize=12)
    plt.yticks(rotation=0)

    # Añadir líneas para cambios de edad
    for i, edad in edades_unicas:
        if i > 0:
            plt.axvline(x=i, color='black', linestyle='--', alpha=0.5)
            plt.text(i+0.2, -0.5, f"Edad {edad}", rotation=90, alpha=0.7)

    plt.tight_layout()
    plt.show()

def generar_reporte_estrategia(resultados, civilizacion=None):
    """
    Genera un reporte completo de la estrategia optimizada

    Args:
        resultados: Diccionario con los resultados
        civilizacion: Nombre de la civilización (opcional)

    Returns:
        Reporte formateado como string
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return "No hay resultados óptimos disponibles para generar un reporte."

    guia_detallada = generar_guia_detallada(resultados)
    guia_por_edad = crear_guia_por_edad(resultados)

    # Crear título del reporte
    titulo = "REPORTE DE ESTRATEGIA OPTIMIZADA PARA AGE OF EMPIRES II"
    if civilizacion:
        titulo += f" - CIVILIZACIÓN: {civilizacion.upper()}"

    # Información general
    info_general = [
        "=" * len(titulo),
        titulo,
        "=" * len(titulo),
        "",
        "RESUMEN EJECUTIVO:",
        "-----------------",
        f"• Valor estratégico optimizado: {resultados['objetivo']:.2f}",
        f"• Total de minutos analizados: {len(resultados['edades'])}",
        f"• Edad final alcanzada: {resultados['edades'][-1]}",
    ]

    # Añadir información sobre momentos de avance
    info_general.extend([
        "",
        "MOMENTOS CLAVE DE AVANCE:",
        "----------------------"
    ])

    for edad, minuto in resultados.get("momentos_avance", {}).items():
        info_general.append(f"• Inicio de avance a Edad {edad}: Minuto {minuto}")

    # Crear sección para cada edad
    secciones_edad = []
    for edad, datos in guia_por_edad.items():
        seccion = [
            "",
            f"ESTRATEGIA PARA LA EDAD {edad.upper()}:",
            "-" * (len(f"ESTRATEGIA PARA LA EDAD {edad.upper()}:") + 2),
            f"Período: {datos['tiempo']}",
            "",
            "Distribución recomendada de ciudadanos:",
            "--------------------------------------"
        ]

        for recurso, cantidad in datos['distribucion_ciudadanos'].items():
            seccion.append(f"• {ICONOS.get(recurso, '')} {recurso.title()}: {cantidad} ciudadanos")

        if datos['edificios_recomendados']:
            seccion.extend([
                "",
                "Edificios a construir:",
                "--------------------"
            ])
            for edificio in datos['edificios_recomendados']:
                seccion.append(f"• {edificio}")

        if datos['unidades_recomendadas']:
            seccion.extend([
                "",
                "Unidades a entrenar:",
                "------------------"
            ])
            for unidad in datos['unidades_recomendadas']:
                seccion.append(f"• {ICONOS.get(unidad['tipo'], '⚔️')} {unidad['tipo'].title()}: {unidad['cantidad']} unidades")

        seccion.extend([
            "",
            "Objetivos clave:",
            "--------------"
        ])
        for objetivo in datos['objetivos_clave']:
            seccion.append(f"• {objetivo}")

        if datos.get('transicion', {}):
            seccion.extend([
                "",
                f"Transición a Edad {datos['transicion']['edad_objetivo']}:",
                "-" * (len(f"Transición a Edad {datos['transicion']['edad_objetivo']}:") + 2),
                f"Momento recomendado: Minuto {datos['transicion']['tiempo_recomendado']}"
            ])

            if datos['transicion'].get('recomendaciones', []):
                seccion.append("Preparación recomendada:")
                for rec in datos['transicion']['recomendaciones']:
                    seccion.append(f"• {rec}")

        secciones_edad.append("\n".join(seccion))

    # Guía minuto a minuto
    guia_minuto = [
        "",
        "GUÍA DETALLADA MINUTO A MINUTO:",
        "=============================",
        ""
    ]

    for paso in guia_detallada:
        minuto = paso['minuto']
        edad = paso['edad']

        guia_minuto.append(f"MINUTO {minuto} - Edad {edad}:")
        guia_minuto.append("-" * (len(f"MINUTO {minuto} - Edad {edad}:") + 2))

        if paso['acciones']:
            guia_minuto.append("Acciones:")
            for accion in paso['acciones']:
                guia_minuto.append(f"  • {accion}")

        if paso['prioridad']:
            guia_minuto.append("Prioridades:")
            for prio in paso['prioridad']:
                guia_minuto.append(f"  • {prio}")

        if paso['consejos']:
            guia_minuto.append("Consejos:")
            for consejo in paso['consejos']:
                guia_minuto.append(f"  • {consejo}")

        guia_minuto.append("")

    # Crear consejos específicos según la civilización
    consejos_civilizacion = []
    if civilizacion:
        # Diccionario con consejos específicos por civilización
        consejos_por_civ = {
            "Francos": [
                "• Aprovecha tu bonificación de caballería para una estrategia agresiva en la Edad de Castillos",
                "• Construye granjas temprano para aprovechar tu bonificación en alimento",
                "• Las mejoras de caballeros y paladines deben ser prioritarias"
            ],
            "Ingleses": [
                "• Defiende con arqueros en la Edad Feudal aprovechando su mayor alcance",
                "• Construye centros urbanos adicionales para aumentar tu producción de arqueros",
                "• La tecnología única Yeomen debe investigarse en cuanto llegues a la Edad de Castillos"
            ],
            "Mongoles": [
                "• Aprovecha la velocidad de tus unidades de caballería para redadas rápidas",
                "• Prioriza el entrenamiento de mangudais en la Edad de Castillos",
                "• Adopta una estrategia móvil, evitando batallas estáticas"
            ],
            "Bizantinos": [
                "• Utiliza tu diversidad de unidades para contrarrestar al enemigo",
                "• Aprovecha las murallas y edificios más resistentes para una defensa sólida",
                "• Los barcos de fuego son excelentes en mapas con agua"
            ],
            "Mayas": [
                "• Aprovecha tu bonificación de recursos al inicio",
                "• Prioriza arqueros de plumed en la Edad de Castillos",
                "• Construye más edificios de forma temprana para aumentar tu población"
            ],
            "Godos": [
                "• Produce infantería en masa, especialmente huscarles en la Edad Imperial",
                "• Aprovecha la bonificación de población con unidades de infantería",
                "• Entrena unidades anti-edificio para asedios rápidos"
            ]
        }

        if civilizacion in consejos_por_civ:
            consejos_civilizacion = [
                "",
                f"CONSEJOS ESPECÍFICOS PARA {civilizacion.upper()}:",
                "----------------------------------------",
            ] + consejos_por_civ[civilizacion]
        else:
            consejos_civilizacion = [
                "",
                "CONSEJOS GENERALES:",
                "-----------------",
                "• Mantén una producción constante de aldeanos en las edades tempranas",
                "• Asegúrate de tener un buen equilibrio entre economía y militar",
                "• Adapta tu estrategia según lo que encuentres en el mapa"
            ]

    # Unir todas las secciones del reporte
    reporte_completo = (
        "\n".join(info_general) + "\n" +
        "\n".join(secciones_edad) + "\n" +
        "\n".join(consejos_civilizacion) + "\n" +
        "\n".join(guia_minuto)
    )

    return reporte_completo

def ejecutar_optimizacion_mejorada(civilizacion=None, params_personalizados=None):
    """
    Función de alto nivel para ejecutar todo el proceso de optimización y visualización

    Args:
        civilizacion: Nombre de la civilización (opcional)
        params_personalizados: Parámetros personalizados para la optimización (opcional)

    Returns:
        Diccionario con todos los resultados
    """
    # Configurar parámetros según la civilización
    if civilizacion and params_personalizados is None:
        params_personalizados = {}

        # Parámetros específicos por civilización
        if civilizacion == "Francos":
            params_personalizados = {
                'RECOLECCION': {"alimento": 30, "madera": 20, "oro": 22, "piedra": 18},
                'VALOR_EDAD': {"Oscura": 1, "Feudal": 2, "Castillos": 8, "Imperial": 10},
                'UTILIDAD_UNIDAD': {
                    "milicia": 1, "arquero": 1.5, "lancero": 1.2,
                    "caballero": 4.5, "mangudai": 3, "paladín": 6
                }
            }
        elif civilizacion == "Ingleses":
            params_personalizados = {
                'RECOLECCION': {"alimento": 25, "madera": 25, "oro": 22, "piedra": 18},
                'VALOR_EDAD': {"Oscura": 1, "Feudal": 4, "Castillos": 6, "Imperial": 9},
                'UTILIDAD_UNIDAD': {
                    "milicia": 1, "arquero": 2.5, "lancero": 1.2,
                    "caballero": 3, "mangudai": 3, "paladín": 4
                }
            }
        elif civilizacion == "Mongoles":
            params_personalizados = {
                'RECOLECCION': {"alimento": 27, "madera": 22, "oro": 25, "piedra": 18},
                'UTILIDAD_UNIDAD': {
                    "milicia": 1, "arquero": 1.5, "lancero": 1.2,
                    "caballero": 3.5, "mangudai": 4.5, "paladín": 5
                }
            }
        elif civilizacion == "Bizantinos":
            params_personalizados = {
                'COSTO_AVANCE': {
                    "Feudal": {"alimento": 450, "madera": 0, "oro": 0, "piedra": 0},
                    "Castillos": {"alimento": 750, "madera": 200, "oro": 0, "piedra": 0},
                    "Imperial": {"alimento": 950, "madera": 750, "oro": 200, "piedra": 0}
                }
            }

    # Ejecutar optimización
    print(f"Ejecutando optimización para {'la civilización ' + civilizacion if civilizacion else 'el juego base'}...")
    resultados = optimizar_aoe2(params_personalizados)

    if resultados:
        # Visualizar resultados
        visualizar_resultados(resultados, civilizacion)

        # Generar reporte
        reporte = generar_reporte_estrategia(resultados, civilizacion)
        print(reporte)

        # También guardar el reporte en un archivo si se desea
        nombre_archivo = f"estrategia_{'_'.join(civilizacion.lower().split()) if civilizacion else 'base'}.txt"
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            f.write(reporte)
        print(f"Reporte guardado en {nombre_archivo}")

        # Opcional: Guardar las visualizaciones en archivos

        return {
            "resultados_optimizacion": resultados,
            "reporte": reporte,
            "guia_detallada": generar_guia_detallada(resultados),
            "guia_por_edad": crear_guia_por_edad(resultados),
            "civilizacion": civilizacion
        }
    else:
        print("La optimización no produjo resultados óptimos.")
        return None

def generar_guia_detallada(resultados):
    """
    Genera una guía detallada minuto a minuto basada en los resultados de la optimización

    Args:
        resultados: Diccionario con los resultados de la optimización

    Returns:
        Lista de diccionarios con guía minuto a minuto
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return []

    guia = []
    recursos = resultados.get("recursos", {})
    ciudadanos = resultados.get("ciudadanos", {})
    edades = resultados.get("edades", [])
    unidades = resultados.get("unidades", {})
    momentos_avance = resultados.get("momentos_avance", {})

    for t in range(len(edades)):
        # Crear entrada para este minuto
        entrada = {
            "minuto": t,
            "edad": edades[t],
            "acciones": [],
            "prioridad": [],
            "consejos": []
        }

        # Determinar acciones basadas en cambios en ciudadanos
        for recurso in ciudadanos:
            # Verificar que el recurso tiene suficientes datos para este tiempo
            valores_recurso = ciudadanos.get(recurso, [])

            # Verificar que hay suficientes elementos en la lista para acceder a t y t-1
            if t > 0 and t < len(valores_recurso) and (t-1) < len(valores_recurso):
                if valores_recurso[t] > valores_recurso[t-1]:
                    diferencia = valores_recurso[t] - valores_recurso[t-1]
                    entrada["acciones"].append(f"Asignar {diferencia} nuevos aldeanos a {recurso}")
                elif valores_recurso[t] < valores_recurso[t-1]:
                    diferencia = valores_recurso[t-1] - valores_recurso[t]
                    entrada["acciones"].append(f"Reasignar {diferencia} aldeanos desde {recurso}")

        # Determinar si es un momento de avance de edad
        for edad, minuto in momentos_avance.items():
            if t == minuto:
                entrada["acciones"].append(f"Iniciar avance a Edad {edad}")
                entrada["prioridad"].append(f"Asegurar recursos para completar el avance a Edad {edad}")

        # Determinar si es un cambio de edad
        if t > 0 and t < len(edades) and (t-1) < len(edades) and edades[t] != edades[t-1]:
            entrada["acciones"].append(f"¡Has alcanzado la Edad {edades[t]}!")

            if edades[t] == "Feudal":
                entrada["prioridad"].append("Construir edificios militares")
                entrada["consejos"].append("Explora el mapa y evalúa la posición enemiga")
            elif edades[t] == "Castillos":
                entrada["prioridad"].append("Construir castillo y expandir economía")
                entrada["consejos"].append("Considera realizar un ataque con unidades de castillo")
            elif edades[t] == "Imperial":
                entrada["prioridad"].append("Investigar tecnologías imperiales y comenzar producción de unidades elite")
                entrada["consejos"].append("Prepárate para un empuje final con unidades avanzadas")

        # Determinar producción de unidades
        for unidad, tiempos in unidades.items():
            for tiempo, cantidad in tiempos:
                if tiempo == t and cantidad > 0:
                    entrada["acciones"].append(f"Producir {cantidad} unidades de {unidad}")

        # Añadir consejos generales según la etapa del juego
        if t < 10:
            entrada["consejos"].append("Mantén una producción constante de aldeanos")
        elif 10 <= t < 20:
            entrada["consejos"].append("Equilibra tu economía y comienza a desarrollar tu ejército")
        elif t >= 20:
            entrada["consejos"].append("Mantén la presión en el enemigo y sigue expandiendo tu economía")

        guia.append(entrada)

    return guia
def crear_guia_por_edad(resultados):
    """
    Crea una guía estratégica organizada por edades

    Args:
        resultados: Diccionario con los resultados de la optimización

    Returns:
        Diccionario con estrategias por edad
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return {}

    edades = resultados.get("edades", [])
    ciudadanos = resultados.get("ciudadanos", {})
    recursos = resultados.get("recursos", {})
    unidades = resultados.get("unidades", {})
    momentos_avance = resultados.get("momentos_avance", {})

    # Identificar rangos de tiempo para cada edad
    cambios_edad = {}
    edad_actual = None

    for t, edad in enumerate(edades):
        if edad != edad_actual:
            if edad not in cambios_edad:
                cambios_edad[edad] = {"inicio": t}
            if edad_actual is not None:
                cambios_edad[edad_actual]["fin"] = t - 1
            edad_actual = edad

    # Establecer el final de la última edad
    if edad_actual is not None:
        cambios_edad[edad_actual]["fin"] = len(edades) - 1

    # Crear guía para cada edad
    guia_edades = {}

    for edad, rango in cambios_edad.items():
        inicio, fin = rango["inicio"], rango["fin"]

        # Calcular distribución promedio de ciudadanos para esta edad
        dist_ciudadanos = {}
        for recurso in ciudadanos:
            # Verificar que los índices estén dentro del rango disponible
            valores = []
            for t in range(inicio, fin + 1):
                if t < len(ciudadanos[recurso]):  # Verificar que el índice sea válido
                    valores.append(ciudadanos[recurso][t])

            dist_ciudadanos[recurso] = int(sum(valores) / len(valores)) if valores else 0

        # Determinar unidades recomendadas para esta edad
        unidades_recomendadas = []
        for unidad, tiempos in unidades.items():
            cantidad_total = sum(cantidad for t, cantidad in tiempos if inicio <= t <= fin)
            if cantidad_total > 0:
                unidades_recomendadas.append({"tipo": unidad, "cantidad": cantidad_total})

        # Determinar edificios recomendados para esta edad
        edificios_recomendados = []
        if edad == "Feudal":
            edificios_recomendados = ["Arquería", "Cuartel", "Torre de vigilancia", "Herrería"]
        elif edad == "Castillos":
            edificios_recomendados = ["Castillo", "Taller de asedio", "Monasterio", "Mercado"]
        elif edad == "Imperial":
            edificios_recomendados = ["Universidad", "Taller de artillería", "Maravilla"]

        # Determinar objetivos clave para esta edad
        objetivos_clave = []
        if edad == "Oscura":
            objetivos_clave = [
                "Construir una economía sólida",
                "Explorar el mapa",
                "Prepararse para el avance a Feudal"
            ]
        elif edad == "Feudal":
            objetivos_clave = [
                "Iniciar producción militar básica",
                "Expandir la economía",
                "Defender de ataques tempranos",
                "Preparar avance a Castillos"
            ]
        elif edad == "Castillos":
            objetivos_clave = [
                "Establecer una presencia militar fuerte",
                "Expandir tu territorio",
                "Iniciar producción de unidades especiales",
                "Evaluar estrategia para la Edad Imperial"
            ]
        elif edad == "Imperial":
            objetivos_clave = [
                "Producir unidades avanzadas",
                "Investigar tecnologías únicas",
                "Lanzar ataques coordinados",
                "Asegurar recursos estratégicos"
            ]

        # Determinar información de transición a la siguiente edad
        transicion = {}
        for siguiente_edad, tiempo in momentos_avance.items():
            if inicio <= tiempo <= fin:
                requisitos_transicion = []
                for requisito in ["alimento", "oro", "madera", "piedra"]:
                    if requisito in recursos and tiempo < len(recursos[requisito]) and recursos[requisito][tiempo] > 0:
                        requisitos_transicion.append(f"Asegura {requisito} suficiente para la transición")

                transicion = {
                    "edad_objetivo": siguiente_edad,
                    "tiempo_recomendado": tiempo,
                    "recomendaciones": requisitos_transicion
                }
                break

        # Crear entrada para esta edad
        guia_edades[edad] = {
            "tiempo": f"Minutos {inicio} - {fin}",
            "distribucion_ciudadanos": dist_ciudadanos,
            "unidades_recomendadas": unidades_recomendadas,
            "edificios_recomendados": edificios_recomendados,
            "objetivos_clave": objetivos_clave,
            "transicion": transicion
        }

    return guia_edades

# Si este script se ejecuta directamente
if __name__ == "__main__":
    # Ejemplo de uso
    resultado = ejecutar_optimizacion_mejorada("Ingleses")
```
