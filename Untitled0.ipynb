```python
import pulp as pl
import numpy as np
import matplotlib.pyplot as plt
import time
from tabulate import tabulate # Para mostrar tablas formateadas bonitas (instalar con pip install tabulate)

def optimizar_aoe2(params=None):
    """
    Función principal para optimizar la estrategia en Age of Empires II.

    Args:
        params: Diccionario con parámetros personalizados para la simulación.

    Returns:
        Un diccionario con los resultados de la optimización si es exitosa,
        o None si no se encuentra una solución óptima.
    """
    # Usar parámetros por defecto si no se proporcionan
    if params is None:
        params = {}

    # --- Parámetros del modelo con valores por defecto ---
    # Convertir T_MAX a segundos para cálculos más finos si es necesario,
    # pero mantener la interfaz en minutos por simplicidad.
    T_MAX = params.get('T_MAX', 20) # Simulación por defecto hasta 20 minutos
    # MAX_CIUDADANOS AUMENTADO para mayor factibilidad
    MAX_CIUDADANOS = params.get('MAX_CIUDADANOS', 100) # Límite de población (antes 50)
    INICIO_CIUDADANOS = params.get('INICIO_CIUDADANOS', 6) # Inicio con 6 aldeanos
    # RECURSOS INICIALES AUMENTADOS para mayor factibilidad
    INICIO_RECURSOS = params.get('INICIO_RECURSOS',
                                 {"alimento": 500, "madera": 200, "oro": 0, "piedra": 0}) # Recursos iniciales (antes 200 ali, 50 mad)

    # Definiciones básicas
    EDADES = ["Oscura", "Feudal", "Castillos", "Imperial"]
    RECURSOS = ["alimento", "madera", "oro", "piedra"]

    # Unidades y su edificio de creación (se ha refinado la lista)
    UNIDADES = {
        "milicia": "cuartel",
        "arquero": "arquería",
        "lancero": "cuartel",
        "explorador": "establo",
        "caballero": "establo",
        "mangudai": "castillo", # Unidad única mongola
        "monje": "monasterio",
        "ariete": "taller_de_asedio",
        "paladín": "establo", # Mejora del caballero
        "onagro": "taller_de_asedio",
        "alabardero": "cuartel", # Mejora del lancero
        "guerrillero": "arquería", # Unidad genérica de arquería
        "camello": "establo", # Unidad de caballería
        "samurai": "castillo", # Ejemplo de unidad única japonesa
        "ballestero": "arquería" # Ejemplo de mejora de arquero
    }
    todas_unidades = list(UNIDADES.keys())

    # Edificios requeridos por edad y su estructura
    EDIFICIOS_POR_EDAD = {
        "Oscura": ["centro_urbano", "molino", "campamento_maderero", "campamento_minero", "cuartel"],
        "Feudal": ["arquería", "establo", "herrería", "mercado"],
        "Castillos": ["castillo", "monasterio", "taller_de_asedio", "universidad"],
        "Imperial": ["torre_de_bombardeo"] # El taller de asedio ya existe en castillos, pero se considera para unidades imperiales
    }
    # Mapeo inverso para saber a qué edad pertenece cada edificio
    EDIFICIO_A_EDAD = {}
    for edad, edificios_lista in EDIFICIOS_POR_EDAD.items():
        for edificio in edificios_lista:
            EDIFICIO_A_EDAD[edificio] = edad
    all_buildings = list(set([b for sublist in EDIFICIOS_POR_EDAD.values() for b in sublist]))

    # Costos de edificios
    COSTO_EDIFICIO = {
        "centro_urbano": {"madera": 275},
        "molino": {"madera": 100},
        "campamento_maderero": {"madera": 100},
        "campamento_minero": {"madera": 100},
        "cuartel": {"madera": 100},
        "arquería": {"madera": 175},
        "establo": {"madera": 175},
        "herrería": {"madera": 150},
        "mercado": {"madera": 175},
        "castillo": {"piedra": 650},
        "monasterio": {"madera": 175},
        "taller_de_asedio": {"madera": 200},
        "universidad": {"madera": 200, "piedra": 100},
        "torre_de_bombardeo": {"madera": 100, "piedra": 150},
    }

    # Tasas de recolección (por defecto o personalizadas)
    RECOLECCION = params.get('RECOLECCION', {
        "alimento": 30, "madera": 25, "oro": 20, "piedra": 15
    })

    # Costos de avance de edad
    COSTO_AVANCE = params.get('COSTO_AVANCE', {
        "Feudal": {"alimento": 500, "oro": 0, "madera": 0, "piedra": 0},
        "Castillos": {"alimento": 800, "oro": 200, "madera": 0, "piedra": 0},
        "Imperial": {"alimento": 1000, "oro": 800, "madera": 0, "piedra": 0}
    })

    # Tiempo para avanzar (en minutos)
    TIEMPO_AVANCE = params.get('TIEMPO_AVANCE', {
        "Feudal": 2, "Castillos": 2.5, "Imperial": 3.5
    })

    # Valor estratégico de cada edad
    VALOR_EDAD = params.get('VALOR_EDAD', {
        "Oscura": 1, "Feudal": 3, "Castillos": 6, "Imperial": 10
    })

    # Costos de unidades (más detallados)
    COSTO_UNIDAD = params.get('COSTO_UNIDAD', {
        "milicia": {"alimento": 60, "oro": 20}, "arquero": {"madera": 25, "oro": 45},
        "lancero": {"alimento": 35, "madera": 25}, "explorador": {"alimento": 80},
        "caballero": {"alimento": 60, "oro": 75}, "mangudai": {"madera": 55, "oro": 65},
        "monje": {"oro": 100}, "ariete": {"madera": 160, "oro": 75},
        "paladín": {"alimento": 120, "oro": 75}, "onagro": {"madera": 180, "oro": 150},
        "alabardero": {"alimento": 35, "madera": 25}, "guerrillero": {"madera": 35, "oro": 20},
        "camello": {"alimento": 55, "oro": 60}, "samurai": {"alimento": 60, "oro": 30},
        "ballestero": {"madera": 25, "oro": 45} # Mismo costo que arquero base por simplicidad
    })

    # Utilidad de unidades (ajustada para equilibrio)
    UTILIDAD_UNIDAD = params.get('UTILIDAD_UNIDAD', {
        "milicia": 1, "arquero": 1.5, "lancero": 1.2, "explorador": 1.8,
        "caballero": 3.5, "mangudai": 3, "monje": 2, "ariete": 2.5,
        "paladín": 5, "onagro": 4, "alabardero": 1.3, "guerrillero": 1.4,
        "camello": 3.2, "samurai": 2.5, "ballestero": 2.0
    })

    COSTO_CIUDADANO = params.get('COSTO_CIUDADANO', {"alimento": 50})

    # Filtrar unidades y edificios si la civilización no los tiene (utilidad 0 o no se pueden construir)
    # Esto es crucial para civilizaciones como los aztecas que no tienen caballería.
    unidades_disponibles = [u for u in todas_unidades if UTILIDAD_UNIDAD.get(u, 0) > 0]
    edificios_disponibles = [b for b in all_buildings if b in COSTO_EDIFICIO]


    # Crear problema de optimización
    prob = pl.LpProblem("Age_of_Empires_II_Optimization", pl.LpMaximize)

    # Rangos para iteración
    tiempo = range(1, T_MAX + 1) # Minuto 1 a T_MAX
    tiempo_completo = range(T_MAX + 1) # Incluye t=0

    # ---- VARIABLES DE DECISIÓN ----
    # `ciudadanos[r, t]` : Número de ciudadanos asignados al recurso `r` en el minuto `t`.
    ciudadanos = pl.LpVariable.dicts("ciudadanos", [(r, t) for r in RECURSOS for t in tiempo], lowBound=0, cat='Integer')
    # `recursos[r, t]` : Cantidad de recurso `r` disponible al final del minuto `t`.
    recursos = pl.LpVariable.dicts("recursos", [(r, t) for r in RECURSOS for t in tiempo_completo], lowBound=0, cat='Continuous')
    # `avance[e, t]` : Binaria, 1 si se inicia el avance a la edad `e` en el minuto `t`.
    avance = pl.LpVariable.dicts("avance", [(e, t) for e in EDADES[1:] for t in tiempo], cat='Binary')
    # `en_edad[e, t]` : Binaria, 1 si el jugador está en la edad `e` en el minuto `t`.
    en_edad = pl.LpVariable.dicts("en_edad", [(e, t) for e in EDADES for t in tiempo_completo], cat='Binary')
    # `crear_ciudadano[t]` : Número de ciudadanos creados en el minuto `t`.
    crear_ciudadano = pl.LpVariable.dicts("crear_ciudadano", [t for t in tiempo], lowBound=0, cat='Integer')
    # `crear_unidad[u, t]` : Número de unidades `u` creadas en el minuto `t`.
    crear_unidad = pl.LpVariable.dicts("crear_unidad", [(u, t) for u in unidades_disponibles for t in tiempo], lowBound=0, cat='Integer')
    # `construir_edificio[b, t]` : Binaria, 1 si el edificio `b` se construye en el minuto `t`.
    construir_edificio = pl.LpVariable.dicts("construir_edificio", [(b, t) for b in edificios_disponibles for t in tiempo], cat='Binary')
    # `edificio_construido_acumulado[b, t]` : Binaria, 1 si el edificio `b` ha sido construido antes o en el minuto `t`.
    edificio_construido_acumulado = pl.LpVariable.dicts("edificio_construido_acumulado",
                                                      [(b, t) for b in edificios_disponibles for t in tiempo_completo], cat='Binary')

    # ---- CONDICIONES INICIALES ----
    # En t=0, estamos en Edad Oscura y los recursos iniciales.
    prob += en_edad["Oscura", 0] == 1, "Inicio_en_Edad_Oscura"
    for e in EDADES[1:]:
        prob += en_edad[e, 0] == 0, f"No_en_{e}_al_inicio"

    for r in RECURSOS:
        prob += recursos[r, 0] == INICIO_RECURSOS.get(r, 0), f"Recursos_Iniciales_{r}"

    # Ciudadanos iniciales asignados al alimento al principio del minuto 1
    # Esto es crucial para que la recolección empiece.
    prob += pl.lpSum(ciudadanos[r, 1] for r in RECURSOS) == INICIO_CIUDADANOS, "Ciudadanos_iniciales_total"
    # Una forma de forzar la asignación inicial al alimento es así:
    prob += ciudadanos["alimento", 1] == INICIO_CIUDADANOS, "Aldeanos_iniciales_al_alimento"
    prob += ciudadanos["madera", 1] == 0, "Aldeanos_iniciales_madera_0"
    prob += ciudadanos["oro", 1] == 0, "Aldeanos_iniciales_oro_0"
    prob += ciudadanos["piedra", 1] == 0, "Aldeanos_iniciales_piedra_0"


    # Ningún edificio construido al inicio
    for b in edificios_disponibles:
        prob += edificio_construido_acumulado[b, 0] == 0, f"No_hay_{b}_construido_al_inicio"
    # El centro urbano siempre existe al inicio
    if "centro_urbano" in edificios_disponibles:
        prob += edificio_construido_acumulado["centro_urbano", 0] == 1, "Centro_Urbano_existe_al_inicio"


    # ---- FUNCIÓN OBJETIVO ----
    # Maximizar valor estratégico (suma ponderada de edad alcanzada y unidades creadas)
    objetivo = 0
    for e in EDADES:
        for t in tiempo:
            objetivo += VALOR_EDAD[e] * en_edad[e, t]

    for u in unidades_disponibles:
        for t in tiempo:
            objetivo += UTILIDAD_UNIDAD[u] * crear_unidad[u, t]

    # Incentivar la construcción temprana de edificios - ponderación reducida para mayor factibilidad
    for b in edificios_disponibles:
        for t in tiempo:
            objetivo += (T_MAX - t + 1) * 0.005 * construir_edificio[b, t] # Reducido el factor (antes 0.01)

    prob += objetivo, "Maximizar_Valor_Estrategico"

    # ---- RESTRICCIONES ----

    # 1. Balance de recursos
    for r in RECURSOS:
        for t in tiempo:
            gasto_avance = pl.lpSum(COSTO_AVANCE[e].get(r, 0) * avance[e, t] for e in EDADES[1:])
            gasto_unidades = pl.lpSum(COSTO_UNIDAD[u].get(r, 0) * crear_unidad[u, t] for u in unidades_disponibles)
            gasto_edificios = pl.lpSum(COSTO_EDIFICIO[b].get(r, 0) * construir_edificio[b, t] for b in edificios_disponibles)
            gasto_ciudadanos = COSTO_CIUDADANO.get(r, 0) * crear_ciudadano[t]

            prob += recursos[r, t] == (recursos[r, t-1]
                                       + RECOLECCION[r] * ciudadanos[r, t]
                                       - gasto_avance
                                       - gasto_unidades
                                       - gasto_edificios
                                       - gasto_ciudadanos), f"Balance_de_Recursos_{r}_en_t={t}"

    # 2. Límite de ciudadanos y población total
    for t in tiempo:
        prob += pl.lpSum(ciudadanos[r, t] for r in RECURSOS) <= MAX_CIUDADANOS, f"Limite_Ciudadanos_en_t={t}"
        # La suma de ciudadanos en t+1 es la de t más los nuevos creados
        if t < T_MAX:
             prob += pl.lpSum(ciudadanos[r, t+1] for r in RECURSOS) == pl.lpSum(ciudadanos[r, t] for r in RECURSOS) + crear_ciudadano[t], f"Poblacion_de_ciudadanos_en_t={t+1}"

        # Un centro urbano es necesario para crear ciudadanos
        # Ajuste para permitir la creación de aldeanos desde t=1 si el CU existe en t=0
        if "centro_urbano" in edificios_disponibles:
            # Si t-1 es 0, edificio_construido_acumulado["centro_urbano", 0] es 1
            prob += crear_ciudadano[t] <= 100 * edificio_construido_acumulado["centro_urbano", max(0, t-1)], f"Crear_ciudadano_requiere_Centro_Urbano_en_t={t}"
        prob += crear_ciudadano[t] * COSTO_CIUDADANO["alimento"] <= recursos["alimento", t-1], f"Costo_de_ciudadano_en_t={t}"

    # 3. Requisitos de avance de edad (recursos y edificios)
    for idx, e in enumerate(EDADES[1:]):
        edad_anterior = EDADES[idx]
        for t in tiempo:
            # Solo se puede iniciar el avance si se está en la edad anterior
            prob += avance[e, t] <= en_edad[edad_anterior, t-1], f"Solo_se_puede_avanzar_de_{edad_anterior}_a_{e}_en_t={t}"

            # Requisito de recursos
            for r in RECURSOS:
                if COSTO_AVANCE[e].get(r, 0) > 0:
                    prob += recursos[r, t-1] >= COSTO_AVANCE[e][r] * avance[e, t], f"Recursos_para_avanzar_a_{e}_en_t={t}_{r}"

            # Requisito de edificios (2 edificios de la edad anterior, excepto para Edad Oscura a Feudal)
            if edad_anterior != "Oscura":
                edificios_edad_anterior = EDIFICIOS_POR_EDAD.get(edad_anterior, [])
                # Filtrar solo los edificios realmente disponibles para la civilización
                edificios_edad_anterior_disp = [b for b in edificios_edad_anterior if b in edificios_disponibles]

                if edificios_edad_anterior_disp:
                    prob += pl.lpSum(edificio_construido_acumulado[b, t-1] for b in edificios_edad_anterior_disp) >= 2 * avance[e, t], f"Edificios_para_avanzar_a_{e}_en_t={t}"
            else: # De Oscura a Feudal, se requieren dos edificios básicos (ej. cuartel y molino/camp.mad.)
                # CORRECCIÓN: Se requieren 2 edificios *adicionales* al Centro Urbano.
                # Asumimos que el centro urbano ya está en t=0, entonces solo necesitamos 2 de la lista.
                req_buildings_feudal = []
                if "cuartel" in edificios_disponibles: req_buildings_feudal.append("cuartel")
                if "molino" in edificios_disponibles: req_buildings_feudal.append("molino")
                if "campamento_maderero" in edificios_disponibles: req_buildings_feudal.append("campamento_maderero")
                if "campamento_minero" in edificios_disponibles: req_buildings_feudal.append("campamento_minero")

                # Se necesitan al menos 2 edificios de los de Edad Oscura (aparte del CU)
                if len(req_buildings_feudal) >= 2:
                    prob += pl.lpSum(edificio_construido_acumulado[b, t-1] for b in req_buildings_feudal) >= 2 * avance[e, t], f"Req_edificios_Feudal_en_t={t}"
                elif len(req_buildings_feudal) == 1: # Si solo hay uno elegible, debe ser ese uno
                    prob += edificio_construido_acumulado[req_buildings_feudal[0], t-1] >= 1 * avance[e, t], f"Req_edificios_Feudal_en_t={t}"
                # Si no hay 2 edificios elegibles (más allá del CU) en la Edad Oscura, esto podría causar inviabilidad.
                # Si una civilización tiene menos de 2 edificios económicos + cuartel, este else podría necesitar ajuste.


    # 4. Transición entre edades (una edad por minuto)
    for t in tiempo_completo:
        prob += pl.lpSum(en_edad[e, t] for e in EDADES) == 1, f"Una_sola_edad_por_t={t}"

    for e_idx, e_actual in enumerate(EDADES):
        for t in tiempo_completo:
            if t > 0:
                # La edad actual se mantiene si no se ha iniciado el avance a la siguiente edad
                # Esta restricción asegura que si se empieza a avanzar a una edad superior,
                # no se puede estar en la edad anterior si el tiempo de avance ha pasado.
                prob += en_edad[e_actual, t] <= en_edad[e_actual, t-1] + pl.lpSum(
                    avance[edad_transicion, tau]
                    for edad_transicion_idx, edad_transicion in enumerate(EDADES)
                    if edad_transicion_idx > e_idx # Para edades superiores a la actual
                    for tau in range(max(1, t - int(TIEMPO_AVANCE.get(edad_transicion, 0))), t + 1) # Si se inició avance en tau y ya ha pasado el tiempo
                    if e_actual == EDADES[edad_transicion_idx - 1] # Si estamos en la edad anterior a la que se avanza
                ), f"Transicion_edad_{e_actual}_mantener_{t}_de_anterior_{e_idx}"

                # Si se avanza a una edad E en el minuto `tau`, se estará en la edad E en `tau + TIEMPO_AVANCE[E]`
                # Asegurarse que la edad actual sea la siguiente a la que se avanza
                if e_idx > 0:
                    for tau in range(1, t): # Minuto de inicio de avance
                        if tau + int(TIEMPO_AVANCE[e_actual]) == t: # Si el avance completa en este minuto 't'
                            prob += en_edad[e_actual, t] >= avance[e_actual, tau], f"Activar_{e_actual}_por_avance_en_{tau}_para_t={t}"

                # Si se avanza a una edad superior, se sale de la edad actual.
                if e_idx < len(EDADES) - 1:
                    siguiente_edad = EDADES[e_idx + 1]
                    for tau in range(1, t + 1):
                        if tau + int(TIEMPO_AVANCE.get(siguiente_edad, 0)) <= t and t > 0: # Si ya se debería haber completado el avance
                             # LÍNEA CORREGIDA PARA NOMBRES ÚNICOS
                             prob += en_edad[e_actual, t] + avance[siguiente_edad, tau] <= 1, f"Salir_de_{e_actual}_si_avanzó_a_{siguiente_edad}_en_t={t}_desde_avance_en_{tau}"

            # Una vez que se alcanza una edad, no se puede "des-alcanzar" (monotonicidad)
            if t > 0:
                prob += en_edad[e_actual, t] >= en_edad[e_actual, t-1], f"Mantener_edad_{e_actual}_en_t={t}_por_monotonicidad"


    # 5. Restricciones para la creación de unidades militares
    for u in unidades_disponibles:
        edificio_requerido = UNIDADES[u]
        edad_unidad_disponible = EDIFICIO_A_EDAD.get(edificio_requerido, "Oscura") # La edad en la que el edificio aparece
        edad_idx_unidad = EDADES.index(edad_unidad_disponible)

        for t in tiempo:
            # Las unidades solo se pueden crear si el edificio está construido Y se está en la edad adecuada
            prob += crear_unidad[u, t] <= 100 * edificio_construido_acumulado[edificio_requerido, t-1], \
                    f"Crear_{u}_requiere_{edificio_requerido}_en_t={t}"

            # Asegurarse de que la edad actual sea igual o superior a la edad de la unidad
            suma_edades_apta = pl.lpSum(en_edad[e_check, t-1] for e_check_idx, e_check in enumerate(EDADES) if e_check_idx >= edad_idx_unidad)
            prob += crear_unidad[u, t] <= 100 * suma_edades_apta, f"Crear_{u}_solo_en_edad_apropiada_en_t={t}"

            # Verificar recursos para crear la unidad
            for r in RECURSOS:
                if COSTO_UNIDAD[u].get(r, 0) > 0:
                    prob += crear_unidad[u, t] * COSTO_UNIDAD[u][r] <= recursos[r, t-1], f"Recursos_para_{u}_en_t={t}_{r}"

    # 6. Restricciones para la construcción de edificios
    for b in edificios_disponibles:
        edad_requerida_edificio = EDIFICIO_A_EDAD.get(b, "Oscura") # Edad en la que el edificio es disponible
        edad_idx_edificio = EDADES.index(edad_requerida_edificio)

        for t in tiempo:
            # Un edificio solo se puede construir una vez
            prob += pl.lpSum(construir_edificio[b, tau] for tau in range(1, t+1)) <= 1, f"Edificio_{b}_se_construye_una_vez_en_t={t}"

            # Un edificio solo se puede construir si se está en la edad adecuada o superior
            suma_edades_apta_edificio = pl.lpSum(en_edad[e_check, t-1] for e_check_idx, e_check in enumerate(EDADES) if e_check_idx >= edad_idx_edificio)
            prob += construir_edificio[b, t] <= suma_edades_apta_edificio, f"Construir_{b}_solo_en_edad_apropiada_en_t={t}"

            # Verificar recursos para construir el edificio
            for r in RECURSOS:
                if COSTO_EDIFICIO[b].get(r, 0) > 0:
                    prob += construir_edificio[b, t] * COSTO_EDIFICIO[b][r] <= recursos[r, t-1], f"Recursos_para_{b}_en_t={t}_{r}"

    # 7. Variable acumulativa de edificios construidos
    for b in edificios_disponibles:
        for t in tiempo:
            prob += edificio_construido_acumulado[b, t] == (edificio_construido_acumulado[b, t-1] + construir_edificio[b, t]), \
                    f"Acumulado_{b}_en_t={t}"

    # ---- RESOLVER EL PROBLEMA ----
    print("Resolviendo el problema de optimización...")
    start_time = time.time()

    # Usar el solver con límite de tiempo y más hilos para robustez
    solver = pl.PULP_CBC_CMD(msg=False, timeLimit=300, threads=4) # Aumentado a 5 minutos, 4 hilos
    try:
        prob.solve(solver)
    except Exception as e:
        print(f"Error durante la resolución: {e}")
        print("Esto podría indicar un problema en la definición del modelo o que el solver no pudo ejecutar.")
        return None

    elapsed_time = time.time() - start_time
    print(f"Tiempo de resolución: {elapsed_time:.2f} segundos")
    print(f"Estado del Solver: {pl.LpStatus[prob.status]}")

    if pl.LpStatus[prob.status] != "Optimal":
        print("No se encontró una solución óptima en el tiempo límite o con los parámetros dados.")
        print("Esto puede ocurrir si el modelo es inviable (no hay ninguna solución posible) o si el tiempo límite es muy bajo.")
        print("Considera ajustar los parámetros iniciales o revisar las restricciones si sigue siendo 'Infeasible'.")
        return None

    print(f"Valor óptimo de la función objetivo: {pl.value(prob.objective):.2f}")

    # Recopilar resultados
    resultados = {
        "status": pl.LpStatus[prob.status],
        "objetivo": pl.value(prob.objective),
        "recursos": {},
        "ciudadanos": {},
        "total_ciudadanos_por_tiempo": [],
        "edades": [],
        "unidades": {},
        "edificios": {},
        "momentos_avance": {},
        "INICIO_CIUDADANOS": INICIO_CIUDADANOS # Añadir para el informe
    }

    # Extraer valores de las variables
    for r in RECURSOS:
        resultados["recursos"][r] = [pl.value(recursos[r, t]) for t in tiempo_completo]

    for r in RECURSOS:
        resultados["ciudadanos"][r] = [pl.value(ciudadanos[r, t]) for t in tiempo]

    resultados["total_ciudadanos_por_tiempo"] = [pl.value(pl.lpSum(ciudadanos[r, t] for r in RECURSOS)) for t in tiempo]

    for t in tiempo_completo:
        found_age = False
        for e in EDADES:
            if pl.value(en_edad[e, t]) > 0.5:
                resultados["edades"].append(e)
                found_age = True
                break
        if not found_age:
            resultados["edades"].append("Desconocida")

    for u in unidades_disponibles:
        unidades_por_tiempo = []
        for t in tiempo:
            valor = pl.value(crear_unidad[u, t])
            if valor > 0.5:
                unidades_por_tiempo.append((t, round(valor)))
        if unidades_por_tiempo:
            resultados["unidades"][u] = unidades_por_tiempo

    for b in edificios_disponibles:
        for t in tiempo:
            if pl.value(construir_edificio[b, t]) > 0.5:
                if b not in resultados["edificios"]:
                    resultados["edificios"][b] = []
                resultados["edificios"][b].append(t)

    for e in EDADES[1:]:
        for t in tiempo:
            if pl.value(avance[e, t]) > 0.5:
                resultados["momentos_avance"][e] = t
                break

    return resultados

def generar_guia_detallada(resultados):
    """
    Genera una guía detallada paso a paso basada en los resultados de la optimización.
    Args:
        resultados: Diccionario con los resultados de optimización.
    Returns:
        Lista de diccionarios con instrucciones detalladas por minuto.
    """
    if resultados is None or resultados.get("status") != "Optimal":
        return []

    edades_historial = resultados.get("edades", [])
    ciudadanos_recurso_hist = resultados.get("ciudadanos", {}) # {'alimento': [v1,v2,...], ...}
    unidades_creadas = resultados.get("unidades", {})
    edificios_construidos = resultados.get("edificios", {})

    # Calcular el total de ciudadanos por minuto del minuto 1 al T_MAX
    total_ciudadanos_por_tiempo = []
    if ciudadanos_recurso_hist:
        max_time_idx = len(next(iter(ciudadanos_recurso_hist.values()))) # Asume que todos los recursos tienen la misma longitud
        total_ciudadanos_por_tiempo = [
            sum(ciudadanos_recurso_hist[r][t_idx] for r in ciudadanos_recurso_hist)
            for t_idx in range(max_time_idx)
        ]

    guia = []
    tiempo_max = len(edades_historial) - 1 # Se ajusta a la duración de los 'tiempos' (minutos 1 a T_MAX)

    for t_idx in range(tiempo_max): # Iterar para los minutos 1 a T_MAX
        minuto = t_idx + 1 # El minuto actual (1-indexed)
        acciones = []

        # Asegurarse de que el índice no esté fuera de rango para edades_historial
        current_age = edades_historial[minuto] if minuto < len(edades_historial) else "Desconocida"

        # Transición de edad
        prev_age = edades_historial[minuto-1] if minuto-1 >= 0 and minuto-1 < len(edades_historial) else "N/A"
        if current_age != prev_age and prev_age != "N/A":
            acciones.append(f"¡Has alcanzado la **Edad {current_age}**!")

        # Ciudadanos: creación y distribución
        current_total_villagers = int(total_ciudadanos_por_tiempo[t_idx]) if t_idx < len(total_ciudadanos_por_tiempo) else 0
        prev_total_villagers = int(total_ciudadanos_por_tiempo[t_idx-1]) if t_idx > 0 else resultados['INICIO_CIUDADANOS']

        created_villagers = current_total_villagers - prev_total_villagers
        if created_villagers > 0:
            acciones.append(f"Crea **{created_villagers} aldeano(s)**.")

        # Distribución de aldeanos para el minuto actual
        current_villager_dist = {r: int(ciudadanos_recurso_hist[r][t_idx]) for r in ciudadanos_recurso_hist if t_idx < len(ciudadanos_recurso_hist[r])}

        # Para el minuto 1, solo se indica la distribución inicial
        if minuto == 1:
            if current_villager_dist:
                acciones.append(f"Asigna aldeanos: {', '.join([f'{count} a {res.title()}' for res, count in current_villager_dist.items() if count > 0])}.")
        else: # Para minutos subsiguientes, indicar cambios en la distribución
            prev_villager_dist = {r: int(ciudadanos_recurso_hist[r][t_idx-1]) for r in ciudadanos_recurso_hist if t_idx-1 < len(ciudadanos_recurso_hist[r])}
            changed_assignments = []
            # Detectar cambios solo si los recuentos son diferentes
            for recurso, current_count in current_villager_dist.items():
                prev_count = prev_villager_dist.get(recurso, 0)
                if current_count != prev_count:
                    changed_assignments.append(f"{current_count} a {recurso.title()} (cambio de {current_count - prev_count})")

            if changed_assignments:
                acciones.append(f"Ajusta aldeanos: {', '.join(changed_assignments)}.")


        # Unidades - CORRECCIÓN APLICADA AQUÍ (generar_guia_detallada)
        for unidad, momentos_cantidades in unidades_creadas.items():
            for m, cantidad in momentos_cantidades:
                if m == minuto:
                    nombre_unidad_formateado = unidad.replace('_', ' ').title()
                    acciones.append(f"Crea **{cantidad} {nombre_unidad_formateado}(es)**.")

        # Edificios
        for edificio, momentos in edificios_construidos.items():
            if minuto in momentos:
                acciones.append(f"Construye **{edificio.replace('_', ' ').title()}**.")

        if acciones:
            guia.append({
                "minuto": minuto,
                "edad": current_age,
                "acciones": acciones
            })

    return guia

def crear_guia_por_edad(resultados):
    """
    Crea una guía estratégica organizada por edades.
    Args:
        resultados: Diccionario con los resultados.
    Returns:
        Diccionario con instrucciones por edad.
    """
    if resultados is None or resultados.ado("status") != "Optimal":
        return {}

    edades_historial = resultados.get("edades", [])
    momentos_avance = resultados.get("momentos_avance", {})
    ciudadanos = resultados.get("ciudadanos", {}) # Dict of lists, where list is for minutes 1 to T_MAX
    unidades = resultados.get("unidades", {})
    edificios = resultados.get("edificios", {})
    # recursos_finales = {r: vals[-1] for r, vals in resultados.get("recursos", {}).items()} # Recursos al final de T_MAX

    guia_por_edad = {}

    # Determinar los intervalos de tiempo para cada edad
    intervalos_edad = {}

    # Asegurarse de que edades_historial tenga al menos un elemento para evitar errores de índice
    if not edades_historial:
        return {}

    # Initial state for age tracking
    current_age_tracking = edades_historial[0]
    start_min_age = 1 if len(edades_historial) > 1 else 0 # If only minute 0, start_min is 0

    for t_idx, edad in enumerate(edades_historial):
        if t_idx == 0: # Skip minute 0 for interval calculation as it's just the starting state
            continue

        minuto_real = t_idx # The actual minute (1-indexed)

        if edad != current_age_tracking:
            # Age has changed, close the previous interval and open a new one
            intervalos_edad[current_age_tracking] = (start_min_age, minuto_real - 1)
            current_age_tracking = edad
            start_min_age = minuto_real

    # Add the last age interval, ensuring it ends at T_MAX (length of history - 1)
    if current_age_tracking is not None:
        intervalos_edad[current_age_tracking] = (start_min_age, len(edades_historial) - 1)


    # Create specific guide for each age
    for edad, (inicio, fin) in intervalos_edad.items():
        distribucion_aldeanos_final = {}
        edificios_en_edad = set()
        unidades_en_edad = {}
        siguiente_edad_objetivo = None
        momento_avance_siguiente = None

        # Distribución de aldeanos al final de la edad
        # 'fin' es el último minuto de esa edad. ciudadanos[] está 0-indexado para minuto 1, etc.
        # Por lo tanto, (fin - 1) es el índice correcto para el minuto 'fin'.
        if ciudadanos and fin > 0 and (fin - 1) < len(next(iter(ciudadanos.values()))): # Asegurarse de que ciudadanos no esté vacío y el índice sea válido
            for recurso in ciudadanos:
                distribucion_aldeanos_final[recurso] = int(ciudadanos[recurso][fin - 1])
        else: # Handle case where there are no villager data or fin is 0
            for recurso in ["alimento", "madera", "oro", "piedra"]:
                distribucion_aldeanos_final[recurso] = 0

        # Edificios construidos en esta edad
        for edificio, momentos in edificios.items():
            for momento_construccion in momentos:
                if inicio <= momento_construccion <= fin:
                    edificios_en_edad.add(edificio)

        # Unidades creadas en esta edad - CORRECCIÓN APLICADA AQUÍ (crear_guia_por_edad)
        for unidad, momentos_cantidades in unidades.items():
            for momento_creacion, cantidad in momentos_cantidades:
                if inicio <= momento_creacion <= fin:
                    unidades_en_edad[unidad] = unidades_en_edad.get(unidad, 0) + cantidad

        # Próximo objetivo de edad
        # Encuentra el avance que ocurre después del inicio de la edad actual
        edades_en_orden = ["Oscura", "Feudal", "Castillos", "Imperial"]
        idx_edad_actual = edades_en_orden.index(edad) if edad in edades_en_orden else -1

        for i in range(idx_edad_actual + 1, len(edades_en_orden)):
            e_siguiente = edades_en_orden[i]
            if e_siguiente in momentos_avance:
                momento_inicio_avance = momentos_avance[e_siguiente]
                # Se considera un avance objetivo si se inicia dentro de la duración de esta edad
                # O si es la siguiente edad lógica en el camino y el avance se inicia *después* de esta edad
                if momento_inicio_avance >= inicio and momento_inicio_avance <= fin:
                    siguiente_edad_objetivo = e_siguiente
                    momento_avance_siguiente = momento_inicio_avance + TIEMPO_AVANCE.get(e_siguiente, 0)
                    break
                elif momento_inicio_avance > fin: # If advance starts after this age ends, it's the next goal
                    siguiente_edad_objetivo = e_siguiente
                    momento_avance_siguiente = momento_inicio_avance + TIEMPO_AVANCE.get(e_siguiente, 0)
                    break # Found the very next age to advance to after this one


        guia_por_edad[edad] = {
            "duracion": f"Minuto {inicio} a {fin}",
            "distribucion_aldeanos_final": distribucion_aldeanos_final,
            "edificios_construidos": list(edificios_en_edad),
            # CORRECCIÓN DE F-STRING APLICADA AQUÍ para unidades_creadas
            "unidades_creadas": {u.replace('_', ' ').title(): cant for u, cant in unidades_en_edad.items()},
            "objetivo_avance": f"Avanzar a Edad {siguiente_edad_objetivo} (alcanzada en min. {int(momento_avance_siguiente)})" if siguiente_edad_objetivo else "No se planea un avance de edad en esta simulación."
        }

    return guia_por_edad

def mostrar_resultados_consolidados(resultados):
    """Muestra un resumen consolidado de los resultados de la optimización."""
    if resultados is None or resultados.get("status") != "Optimal":
        print("\nNo hay resultados óptimos para mostrar.")
        return

    print("\n" + "="*50)
    print(" RESUMEN DE ESTRATEGIA ÓPTIMA DE AGE OF EMPIRES II ")
    print("="*50)

    print(f"\nEstado de la Optimización: **{resultados['status']}**")
    print(f"Valor Estratégico Total (Objetivo): **{resultados['objetivo']:.2f}**")
    print(f"Duración de la Simulación: **{len(resultados['edades'])-1} minutos**")

    # Mostrar la evolución de los recursos
    print("\n--- Recursos al Final de la Simulación ---")
    recursos_finales = {r: int(resultados["recursos"][r][-1]) for r in resultados["recursos"]}
    headers_recursos = ["Recurso", "Cantidad Final"]
    data_recursos = [[r.title(), cant] for r, cant in recursos_finales.items()]
    print(tabulate(data_recursos, headers=headers_recursos, tablefmt="fancy_grid"))

    # Mostrar edificios construidos y cuándo
    print("\n--- Edificios Clave Construidos ---")
    if resultados["edificios"]:
        headers_edificios = ["Edificio", "Minuto de Construcción"]
        data_edificios = []
        for edificio, momentos in resultados["edificios"].items():
            data_edificios.append([edificio.replace('_', ' ').title(), ", ".join(map(str, sorted(momentos)))])
        print(tabulate(data_edificios, headers=headers_edificios, tablefmt="fancy_grid"))
    else:
        print("No se construyeron edificios específicos más allá del Centro Urbano inicial.")

    # Mostrar unidades creadas
    print("\n--- Unidades Militares Creadas ---")
    if resultados["unidades"]:
        headers_unidades = ["Unidad", "Cantidad", "Minutos de Creación"]
        data_unidades = []
        # CORRECCIÓN APLICADA AQUÍ (mostrar_resultados_consolidados)
        for unidad, momentos_cantidades in resultados["unidades"].items():
            total_creadas = sum(cantidad for _, cantidad in momentos_cantidades)
            minutos = sorted(list(set(m for m, _ in momentos_cantidades))) # Minutos únicos de creación
            nombre_unidad_formateado = unidad.replace('_', ' ').title()
            data_unidades.append([nombre_unidad_formateado, total_creadas, ", ".join(map(str, minutos))])
        print(tabulate(data_unidades, headers=headers_unidades, tablefmt="fancy_grid"))
    else:
        print("No se crearon unidades militares.")

    # Mostrar el camino de avance de edad
    print("\n--- Avance de Edades ---")
    edad_actual = resultados["edades"][0] # Edad en minuto 0
    camino_edades = [f"Edad {edad_actual}"]

    # Recorrer los avances de edad planificados
    # Los `momentos_avance` guardan el minuto en que se *inicia* el avance.
    # Necesitamos saber cuándo se *alcanza* la edad.

    # Mapeo de tiempos de avance desde los parámetros originales
    TIEMPO_AVANCE_DICT = { # Renombrado para evitar conflicto con TIEMPO_AVANCE de params
        "Feudal": 2, "Castillos": 2.5, "Imperial": 3.5
    }

    avances_ordenados = sorted(resultados["momentos_avance"].items(), key=lambda item: item[1])

    for edad_avanzada, minuto_inicio_avance in avances_ordenados:
        tiempo_requerido = TIEMPO_AVANCE_DICT.get(edad_avanzada, 0)
        minuto_llegada = minuto_inicio_avance + int(tiempo_requerido)
        camino_edades.append(f"-> Edad {edad_avanzada} (alcanzada en min. {minuto_llegada})")

    print(" ".join(camino_edades))

    # Gráfico de recursos a lo largo del tiempo
    print("\nGenerando gráfico de recursos (se abrirá en una nueva ventana)...")
    plt.figure(figsize=(12, 6))
    for r in resultados["recursos"]:
        plt.plot(range(len(resultados["recursos"][r])), resultados["recursos"][r], label=r.title())
    plt.xlabel("Minuto de Simulación")
    plt.ylabel("Cantidad de Recurso")
    plt.title("Evolución de Recursos a lo largo del Tiempo")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Gráfico de aldeanos por recurso
    print("Generando gráfico de distribución de aldeanos (se abrirá en una nueva ventana)...")
    plt.figure(figsize=(12, 6))
    # Asegurarse de que el array numpy tenga la longitud correcta
    # `resultados["ciudadanos"]` contiene listas para cada recurso desde el minuto 1 a T_MAX
    # por lo que el rango de tiempo debe ser `range(1, T_MAX + 1)`

    # Determinar la longitud de los datos de ciudadanos
    if resultados["ciudadanos"] and next(iter(resultados["ciudadanos"].values()), None) is not None:
        num_minutos_ciudadanos = len(next(iter(resultados["ciudadanos"].values())))
    else:
        num_minutos_ciudadanos = 0 # No hay datos de ciudadanos para graficar

    if num_minutos_ciudadanos > 0:
        bottom_val = np.zeros(num_minutos_ciudadanos)
        for r in RECURSOS: # Asegurarse de iterar sobre los recursos en orden consistente
            if r in resultados["ciudadanos"]:
                # Asegurarse de que los datos de ciudadanos tengan la misma longitud para la gráfica de pila
                data_for_plot = resultados["ciudadanos"][r][:num_minutos_ciudadanos]
                plt.bar(range(1, num_minutos_ciudadanos + 1), data_for_plot, bottom=bottom_val, label=r.title())
                bottom_val += np.array(data_for_plot)

        # Añadir línea para el total de ciudadanos si está disponible
        if resultados["total_ciudadanos_por_tiempo"]:
            plt.plot(range(1, len(resultados["total_ciudadanos_por_tiempo"]) + 1), resultados["total_ciudadanos_por_tiempo"],
                     color='black', linestyle='--', linewidth=2, label='Total Aldeanos')

        plt.xlabel("Minuto de Simulación")
        plt.ylabel("Número de Ciudadanos")
        plt.title("Distribución de Ciudadanos por Recurso")
        plt.legend()
        plt.grid(True, axis='y')
        plt.show()
    else:
        print("No hay datos de distribución de aldeanos para generar el gráfico.")

    print("\n" + "="*50)
    print(" FIN DEL RESUMEN ")
    print("="*50)

def menu_personalizacion():
    """
    Permite al usuario personalizar los parámetros de la simulación.
    Retorna un diccionario con los parámetros personalizados.
    """
    params = {}
    print("\n--- Personalización de la Simulación ---")
    print("Deja en blanco para usar el valor por defecto.")

    try:
        t_max_input = input("Duración máxima de la simulación en minutos (Ej: 20): ").strip()
        if t_max_input:
            params['T_MAX'] = int(t_max_input)
            if params['T_MAX'] <= 0:
                print("La duración debe ser un número positivo. Usando valor por defecto (20).")
                del params['T_MAX']
    except ValueError:
        print("Entrada inválida. Usando valor por defecto para la duración.")

    try:
        max_ciu_input = input("Límite máximo de ciudadanos (Ej: 100): ").strip() # Default en la pregunta
        if max_ciu_input:
            params['MAX_CIUDADANOS'] = int(max_ciu_input)
            if params['MAX_CIUDADANOS'] <= 0:
                print("El límite de ciudadanos debe ser un número positivo. Usando valor por defecto (100).")
                del params['MAX_CIUDADANOS']
    except ValueError:
        print("Entrada inválida. Usando valor por defecto para el límite de ciudadanos.")

    print("\n--- Recursos Iniciales (deja en blanco para usar por defecto) ---")
    recursos_iniciales = {}
    default_recursos = {"alimento": 500, "madera": 200, "oro": 0, "piedra": 0} # Default en el código
    for res, default_val in default_recursos.items():
        try:
            val_input = input(f"Cantidad inicial de {res.title()} (Defecto: {default_val}): ").strip()
            if val_input:
                recursos_iniciales[res] = int(val_input)
                if recursos_iniciales[res] < 0:
                    print(f"La cantidad de {res.title()} no puede ser negativa. Usando valor por defecto.")
                    recursos_iniciales[res] = default_val
            else:
                recursos_iniciales[res] = default_val
        except ValueError:
            print(f"Entrada inválida. Usando valor por defecto para {res.title()}.")
            recursos_iniciales[res] = default_val
    params['INICIO_RECURSOS'] = recursos_iniciales

    print("\n--- Tasas de Recolección por Recurso (por aldeano/minuto) ---")
    recoleccion = {}
    default_recoleccion = {"alimento": 30, "madera": 25, "oro": 20, "piedra": 15}
    for res, default_val in default_recoleccion.items():
        try:
            val_input = input(f"Tasa de {res.title()} (Defecto: {default_val}): ").strip()
            if val_input:
                recoleccion[res] = float(val_input)
                if recoleccion[res] < 0:
                    print(f"La tasa de {res.title()} no puede ser negativa. Usando valor por defecto.")
                    recoleccion[res] = default_val
            else:
                recoleccion[res] = default_val
        except ValueError:
            print(f"Entrada inválida. Usando valor por defecto para {res.title()}.")
            recoleccion[res] = default_val
    params['RECOLECCION'] = recoleccion

    print("\n--- Valores de Utilidad de Unidades (Influencia en el objetivo) ---")
    utilidad_unidad = {}
    # Usamos un valor de utilidad base para las unidades existentes en el modelo
    # Esto asegura que todas las unidades consideradas en el modelo sean manejadas.
    default_utilidad = {
        "milicia": 1, "arquero": 1.5, "lancero": 1.2, "explorador": 1.8,
        "caballero": 3.5, "mangudai": 3, "monje": 2, "ariete": 2.5,
        "paladín": 5, "onagro": 4, "alabardero": 1.3, "guerrillero": 1.4,
        "camello": 3.2, "samurai": 2.5, "ballestero": 2.0
    }

    for unidad, default_val in default_utilidad.items():
        try:
            val_input = input(f"Utilidad para {unidad.replace('_', ' ').title()} (Defecto: {default_val}): ").strip()
            if val_input:
                utilidad_unidad[unidad] = float(val_input)
                if utilidad_unidad[unidad] < 0:
                    print(f"La utilidad no puede ser negativa. Usando valor por defecto.")
                    utilidad_unidad[unidad] = default_val
            else:
                utilidad_unidad[unidad] = default_val
        except ValueError:
            print(f"Entrada inválida. Usando valor por defecto para {unidad.replace('_', ' ').title()}.")
            utilidad_unidad[unidad] = default_val
    params['UTILIDAD_UNIDAD'] = utilidad_unidad

    print("\n--- Valores Estratégicos de Edades (Influencia en el objetivo) ---")
    valor_edad = {}
    default_valor_edad = {"Oscura": 1, "Feudal": 3, "Castillos": 6, "Imperial": 10}
    for edad, default_val in default_valor_edad.items():
        try:
            val_input = input(f"Valor estratégico para Edad {edad} (Defecto: {default_val}): ").strip()
            if val_input:
                valor_edad[edad] = float(val_input)
                if valor_edad[edad] < 0:
                    print(f"El valor no puede ser negativo. Usando valor por defecto.")
                    valor_edad[edad] = default_val
            else:
                valor_edad[edad] = default_val
        except ValueError:
            print(f"Entrada inválida. Usando valor por defecto para Edad {edad}.")
            valor_edad[edad] = default_val
    params['VALOR_EDAD'] = valor_edad


    print("\n--- Costos de Unidades (modificar solo si cambias el balance del juego) ---")
    costo_unidad_personalizado = {}
    # Recorrer las unidades definidas en el modelo para ofrecer personalización
    for unidad in UNIDADES.keys():
        current_costs = {}
        print(f"Costo de {unidad.replace('_', ' ').title()}:")
        for recurso in RECURSOS:
            # Get default cost from the COSTO_UNIDAD defined globally
            default_cost = globals()['COSTO_UNIDAD'].get(unidad, {}).get(recurso, 0)
            try:
                val_input = input(f"  {recurso.title()} (Defecto: {default_cost}): ").strip()
                if val_input:
                    current_costs[recurso] = int(val_input)
                    if current_costs[recurso] < 0:
                        print(f"El costo de {recurso.title()} no puede ser negativo. Usando valor por defecto.")
                        current_costs[recurso] = default_cost
                else:
                    current_costs[recurso] = default_cost
            except ValueError:
                print(f"Entrada inválida. Usando valor por defecto para {recurso.title()}.")
                current_costs[recurso] = default_cost
        costo_unidad_personalizado[unidad] = current_costs
    params['COSTO_UNIDAD'] = costo_unidad_personalizado


    print("\n--- Costos de Avance de Edad (modificar solo si cambias el balance del juego) ---")
    costo_avance_personalizado = {}
    for edad in ["Feudal", "Castillos", "Imperial"]:
        current_costs = {}
        print(f"Costo para avanzar a Edad {edad}:")
        for recurso in RECURSOS:
            # Get default cost from the COSTO_AVANCE defined globally
            default_cost = globals()['COSTO_AVANCE'].get(edad, {}).get(recurso, 0)
            try:
                val_input = input(f"  {recurso.title()} (Defecto: {default_cost}): ").strip()
                if val_input:
                    current_costs[recurso] = int(val_input)
                    if current_costs[recurso] < 0:
                        print(f"El costo de {recurso.title()} no puede ser negativo. Usando valor por defecto.")
                        current_costs[recurso] = default_cost
                else:
                    current_costs[recurso] = default_cost
            except ValueError:
                print(f"Entrada inválida. Usando valor por defecto para {recurso.title()}.")
                current_costs[recurso] = default_cost
        costo_avance_personalizado[edad] = current_costs
    params['COSTO_AVANCE'] = costo_avance_personalizado

    return params

def main():
    print("Bienvenido al Optimizador de Estrategias de Age of Empires II")
    print("Este programa te ayudará a encontrar una estrategia óptima para tus primeros minutos de juego.")

    while True:
        print("\n--- Menú Principal ---")
        print("1. Ejecutar optimización con parámetros por defecto (20 minutos, 100 aldeanos, recursos mejorados)")
        print("2. Personalizar parámetros de la simulación")
        print("3. Salir")

        opcion = input("Selecciona una opción: ").strip()

        if opcion == '1':
            resultados = optimizar_aoe2() # Usa los nuevos valores por defecto
            if resultados:
                mostrar_resultados_consolidados(resultados)
                guia_detallada = generar_guia_detallada(resultados)
                if guia_detallada:
                    print("\n--- Guía Detallada Minuto a Minuto ---")
                    for entrada in guia_detallada:
                        print(f"\n**Minuto {entrada['minuto']}** (Edad: {entrada['edad']}):")
                        for accion in entrada['acciones']:
                            print(f"  - {accion}")

                    print("\n--- Guía Estratégica por Edad ---")
                    guia_por_edad = crear_guia_por_edad(resultados)
                    for edad, detalles in guia_por_edad.items():
                        print(f"\n**EDAD {edad.upper()}** ({detalles['duracion']}):")
                        print(f"  - Distribución final de aldeanos: {', '.join([f'{int(count)} {res.title()}' for res, count in detalles['distribucion_aldeanos_final'].items() if count > 0])}")
                        if detalles['edificios_construidos']:
                            print(f"  - Edificios clave construidos: {', '.join([b.replace('_', ' ').title() for b in sorted(detalles['edificios_construidos'])])}")
                        if detalles['unidades_creadas']:
                            unidades_formateadas_por_edad = [f"{cant} {u}" for u, cant in detalles['unidades_creadas'].items()]
                            print(f"  - Unidades clave creadas: {', '.join(unidades_formateadas_por_edad)}")
                        print(f"  - Objetivo estratégico principal: {detalles['objetivo_avance']}")

                else:
                    print("\nNo se pudo generar la guía detallada.")
            else:
                print("\nLa optimización no pudo encontrar una solución óptima.")
        elif opcion == '2':
            parametros_personalizados = menu_personalizacion()
            resultados = optimizar_aoe2(parametros_personalizados)
            if resultados:
                mostrar_resultados_consolidados(resultados)
                guia_detallada = generar_guia_detallada(resultados)
                if guia_detallada:
                    print("\n--- Guía Detallada Minuto a Minuto ---")
                    for entrada in guia_detallada:
                        print(f"\n**Minuto {entrada['minuto']}** (Edad: {entrada['edad']}):")
                        for accion in entrada['acciones']:
                            print(f"  - {accion}")

                    print("\n--- Guía Estratégica por Edad ---")
                    guia_por_edad = crear_guia_por_edad(resultados)
                    for edad, detalles in guia_por_edad.items():
                        print(f"\n**EDAD {edad.upper()}** ({detalles['duracion']}):")
                        print(f"  - Distribución final de aldeanos: {', '.join([f'{int(count)} {res.title()}' for res, count in detalles['distribucion_aldeanos_final'].items() if count > 0])}")
                        if detalles['edificios_construidos']:
                            print(f"  - Edificios clave construidos: {', '.join([b.replace('_', ' ').title() for b in sorted(detalles['edificios_construidos'])])}")
                        if detalles['unidades_creadas']:
                            unidades_formateadas_por_edad = [f"{cant} {u}" for u, cant in detalles['unidades_creadas'].items()]
                            print(f"  - Unidades clave creadas: {', '.join(unidades_formateadas_por_edad)}")
                        print(f"  - Objetivo estratégico principal: {detalles['objetivo_avance']}")
                else:
                    print("\nNo se pudo generar la guía detallada con los parámetros personalizados.")
            else:
                print("\nLa optimización no pudo encontrar una solución óptima con los parámetros personalizados.")
        elif opcion == '3':
            print("Saliendo del programa. ¡Hasta la próxima!")
            break
        else:
            print("Opción no válida. Por favor, intenta de nuevo.")

if __name__ == "__main__":
    main()
```
