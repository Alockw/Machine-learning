# <FONT SIZE=5 COLOR="purple"> 1. Introducción a Pytorch </FONT>

- PyTorch es una biblioteca de código abierto para aprendizaje profundo desarrollada por **Facebook’s AI Research Lab (FAIR)** y lanzada oficialmente en 2016.

- Su diseño surge como una evolución de Torch, una biblioteca anterior escrita en Lua, que aunque potente, no era tan flexible ni fácil de integrar con entornos de investigación modernos en Python.

- Facebook creó PyTorch con la intención de ofrecer una herramienta que combinara flexibilidad para la investigación y rendimiento para la producción, dos áreas que tradicionalmente se encontraban separadas.

- Hoy en día, PyTorch es mantenido por la Linux Foundation AI y cuenta con el respaldo de grandes empresas como Meta, Microsoft, NVIDIA y Amazon.

**Características**


| Característica                                     | Descripción                                                                                                                                                  |
| -------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| **Ejecución dinámica (define-by-run)**             | A diferencia de TensorFlow 1.x, PyTorch construye el grafo computacional en tiempo real, lo que permite modificar la arquitectura “sobre la marcha”.         |

| **TorchScript y JIT**                              | Permiten convertir modelos dinámicos en versiones optimizadas y exportables a producción.                                                                    |
| **PyTorch Lightning y Hugging Face**               | Posee un ecosistema sólido de frameworks de alto nivel y modelos preentrenados listos para usar.

## <FONT SIZE=4 COLOR="magenta"> 3.2 Definir y variables y conjuntos de entrenamiento, validación y prueba. </FONT>

```python
# definimos las variables del modelo: respuesta y características.
X = diabetes.drop("Outcome", axis=1).values
y = diabetes["Outcome"].values
```

```python
# Vamos a dividir en tres conjuntos: train, validation y test.
X_temp, X_test, y_temp, y_test = train_test_split(X,
                                                  y,
                                                  test_size=0.2,
                                                  stratify=y,
                                                  random_state=42)

# ahora X_temp (80%) lo partimos en 60% train y 20% val
X_train, X_val, y_train, y_val = train_test_split(X_temp,
                                                  y_temp,
                                                  test_size=0.25,
                                                  stratify=y_temp,
                                                  random_state=42)
# queda 60/20/20
```

## <FONT SIZE=4 COLOR="magenta"> 3.3. Definir la clase de los Datos (tensores) </FONT>

Crear un Dataset y DataLoader de PyTorch.

```python
class DiabetesDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32).view(-1, 1)  # (N,1)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

train_dataset = DiabetesDataset(X_train, y_train)
val_dataset = DiabetesDataset(X_val, y_val)
test_dataset = DiabetesDataset(X_test, y_test)
```



PyTorch entrena con tensores, así que hacemos una clase simple:

**def __init__(self, X, y)**: es el constructor de la clase en Python.

- En Python, cada clase puede tener un método especial llamado __init__. Su función es inicializar (o configurar) los atributos del objeto cuando se crea una nueva instancia.

PyTorch espera que los datos tengan dos métodos obligatorios:

__len__: para saber cuántos ejemplos hay.

__getitem__: para poder pedir el ejemplo i-ésimo.

**class DiabetesDataset(Dataset)**:

  - Estamos creando un dataset propio que PyTorch va a poder entender.

  - Dataset es la clase base de PyTorch.

**self.X = torch.tensor(X, dtype=torch.float32)**

  - Convierte los datos de entrada (X) a un tensor de PyTorch.

  - Se fija que sea float32 porque las redes trabajan en float, no en int ni en float64 (por rendimiento).

**self.y = torch.tensor(y, dtype=torch.float32).view(-1, 1)**

  - Convierte las etiquetas a tensor también.

  - view(-1, 1) es importante: fuerza a que $y$ tenga forma (N, 1).

  - -1 poné la cantidad de filas que haga falta según los datos.

  - 1 solo una sola columna”.

**def __len__(self):**

Devuelve len(self.X), es decir, cuántos ejemplos hay en el dataset.

PyTorch usa esto, por ejemplo, para saber cuántos batches hay por época.

**def __getitem__(self, idx):**

Para que al hacer *dataset[5]* devuelva *(X[5], y[5])*.

## <FONT SIZE=4 COLOR="magenta"> 3.5. Inicializar la clase para la arquitectura del modelo y el método forward </FONT>

Ahora vamos a definir la clase para inicializar la arquitectura del modelo

```python
class DiabetesModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(8, 32),    # 8 variables y 32 neuronas
            nn.ReLU(),
            nn.Linear(32, 16),   # ingresan 32 de la anterior y salen 16 a la siguiente
            nn.ReLU(),
            nn.Linear(16, 1)     # salida 1 logit porque es clasificación binaria
        )

    def forward(self, x):
        return self.net(x)

model = DiabetesModel()
```

- **nn.Module**: es la clase base de todos los modelos de *Pytorch*. Esto le indica a Pytorch: "haremos una red , va a tener capas y el método forward". Además, *DiabetesModel* será una clase hija de *nn.Module*

- **super()** es una función de Python para llamar el método *__init__* de la clase padre, en este caso, *nn.Module*.

- **self.net = nn.Sequential** arquitectura de la red de manera secuencial, es decir, que se ejecuta uno detrás del otro.

- **nn.Linear(8, 32):** Es una capa densa (totalmente conectada). Entran vectores de tamaño 8 (número de características) y sale un vector de longitud 32.

- **nn:ReLu():** Es la función de activación en la primera capa. Introduce una no linealidad

- **nn.Linear(32,16):** Segunda capa densa, totalmente conectada. Que toma los 32 valores de la capa anterior y los comprime a 16.

- **nn:ReLu():** Otra no linealidad en la segunda capa.

- **nn.Linear(16,1)**. Es la capa final, que reduce los 16 valores a un solo valor.

**Observación importante**: En un problema de clasificación binaria se puede configurar de dos maneras

   - Enviarlo a 1, al final (logit) y luego aplicar sigmoide afuera o en la función de pérdida mediante **BCEWithLogitsLoss**

   - Enviarlo a 2 (dos clases) y usar la función  **CrossEntropyLoss**.

Finalmente,

- **def forward(self, x):** indica como fluyen los datos. En este caso, hacia adelante, mediante el método *forward*

- **model = DiabetesModel():** a esto se le llama *instanciar* la clase. Es decir, ya tenemos listo el modelo para pasarle los datos.

Por ejemplo,

```python
# tomamos un lote de 64 elementos aleatorios del conjunto de datos y 8 variables.
x = torch.randn(64, 8)
# hacemos la predicción del modelo en una iteración del método forward
y_pred = model(x)
y_pred[0:10]
```

esos número se llaman **Logits**, es decir, una confianza no normalizada acerca de la clase a la que puede pertenecer.

- $logit << 0 $ el modelo cree fuertemente que la clase es 0 (negativa: no diabetes)

- $logit \approx 0$ el modelo está inseguro ($50%$ de probabilidad).

- $logit >> 0 $ el modelo cree fuertemente que la clase es 1 (positiva: si diabetes)


Para convertirlo en probabilidad se le pasa la función sigmoide


$$\sigma(x) =  \dfrac{1}{1+e^{-x}}$$

## <FONT SIZE=4 COLOR="magenta"> 3.6. Definir la función de pérdida y el optimizador </FONT>

- Se define la función de pérdida. Para clasificación binaria.

$$nn.BCEWithLogitsLoss()$$

- Se define el optimizador. Es decir, cómo se minimizan los errores.

$$torch.optim.Adam$$

- **Adam** : tipo de optimizador (Adaptive Moment Estimation). Aprende tasas de aprendizaje diferentes para cada parámetro. Muy popular por su rapidez y estabilidad.

- **model.parameters()** : todos los pesos y biases del modelo. PyTorch los registra automáticamente cuando llamamos super().__init__() en la clase definida antes. El optimizador necesita saber qué parámetros actualizar.

- **$lr = 10^{-3}$** : tasa de aprendizaje (learning rate). Indica cuánto cambia cada peso en cada actualización. Es un valor estándar, pero puede ajustarse.

```python
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

## <FONT SIZE=4 COLOR="magenta"> 3.6. Definir el método backward y evaluación en validation </FONT>

- **num_epoch:** el modelo se entrana pasando 30 veces sobre todo el conjunto de entrenamiento.

- **model.train():** Se inicia el modelo en modo entrenamiento.

- **running_loss = 0.0** es una variable que va acumulando la pérdida, en todos los batches de esa iteración. Sirve para después sacar el promedio.

- **X_batch, y_batch in train_loader:** recorre el DataLoader de entrenamiento, es decir, los lotes.

- **optimizer.zero_grad():** Antes de calcular los gradientes de este batch, se limpiam los gradientes acumulados del batch anterior. Es obligatorio hacerlo.

- **outputs = model(X_batch):** pasar el método forward.

- **loss = criterion(outputs, y_batch):** Para calcular la pérdida entre los que dice el modelo (outputs) y el valor real de cada lote (y_batch).

  - aplica sigmoid a outputs
  - compara con y_batch
  - devuelve un escalar de pérdida

- **loss.backward():** Calcula los gradientes de la pérdida respecto a cada parámetro del modelo.

- **optimizer.step():** se usan los gradientes calculados para actualizar los pesos del modelo.

- **running_loss += loss.item() * X_batch.size(0)**. Paso clave de aprendizaje

  - **loss.item()** convierte el tensor escalar de pérdida en un número de Python.
  - Se multiplica por el tamaño del batch (X_batch.size(0)) para tener la pérdida total del batch (no la media).
  - Se acumula en *running_loss* para luego sacar el promedio sobre TODO el dataset.

- **epoch_loss = running_loss / len(train_loader.dataset)**

  - **running_loss** = suma de las pérdidas de todos los batches.
  - **len(train_loader.dataset)** = número total de muestras en el dataset de entrenamiento.
  - Esto da la pérdida media por muestra en esa época.

Ahora, la parte de la evaluación

- **model.eval()**: se evalúa el modelo en cada iteración.

- **all_preds = []** y **all_targets = []** variables para ir guardando las predicciones del conjunto de validación.

- **with torch.no_grad()** para que no calcule gradiente. Porque no vamos a entrenar.

- **for X_batch, y_batch in val_loader:** recorremos el DataLoader

- **logits = model(X_batch)**: calculamos los logits. Método forward.

- **probs = torch.sigmoid(logits)**: convertimos a probabilidades.

- **preds = (probs >= 0.5).float()** para determinar la clase de acuerdo con el umbral.

- **all_preds.extend(preds.cpu().numpy())** y **all_targets.extend(y_batch.cpu().numpy())**. Para pasar los tensores a CPU y convertir a arreglos de numpy para aplicar el *accuracy*

- **acc = accuracy_score(all_targets, all_preds)**: **calcular el accuracy**

```python
from sklearn.metrics import accuracy_score

num_epochs = 100

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * X_batch.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)

    # ---- evaluación ----
    model.eval()
    all_preds = []
    all_targets = []
    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            logits = model(X_batch)
            probs = torch.sigmoid(logits)
            preds = (probs >= 0.5).float()
            all_preds.extend(preds.cpu().numpy())
            all_targets.extend(y_batch.cpu().numpy())

    acc = accuracy_score(all_targets, all_preds)
    print(f"Epoch {epoch+1}/{num_epochs} - loss: {epoch_loss:.4f} - acc: {acc:.4f}")
```

## <FONT SIZE=4 COLOR="magenta"> 3.7. Evaluar en test </FONT>

Aca es similar al código anterior pero ya en test.

```python
model.eval()
all_preds, all_targets = [], []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        logits = model(X_batch)
        probs = torch.sigmoid(logits)
        preds = (probs >= 0.5).float()
        all_preds.extend(preds.cpu().numpy())
        all_targets.extend(y_batch.cpu().numpy())

final_acc = accuracy_score(all_targets, all_preds)
print(f"Final test accuracy: {final_acc:.4f}")
```
