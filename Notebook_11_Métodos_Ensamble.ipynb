# <FONT SIZE=5 COLOR="purple"> 1. Algoritmos de Ensamble </FONT>

- De manera sencilla se puede decir que los algoritmos de ensamble son algoritmos formados por algoritmos más simples.

- Estos algoritmos se unen para formar un algoritmo mejor, más potente y eficiente en problemas de clasificación y regresión.

- Están basados en el principio : **la unión hace la fuerza**

- Hay diferentes formas de ensamblar o unir algoritmos. Entre las más usadas están:

   - **bagging**
   - **boosting**



### <FONT SIZE=5 COLOR="blue"> 1.1 Algoritmos de Bagging </FONT>

- **Bagging** viene de **B**ootstrap **Agg**regat**ing**. El Bootstrap es una técnica estadística para crear subconjuntos de datos tomando muestras con reemplazo del mismo tamaño del conjunto de datos original.

- El principio de este tipo de algoritmos es entrenar varios modelos independientes en subconjuntos aleatorios de datos y luego combinar las predicciones de cada uno:
   
   + **Clasificación:** clase mayoritaria

   + **Regresión:** promedio de los valores

- Se busca reducir la varianza. Es decir, cuando tenemos problema de machine learning se mide cuanto cambian las predicciones si entrenamos el modelo con diferentes muestras de datos en el mismo problema. Un problema que tenga alta varianza es muy sensible al cambio del conjunto de entrenamiento. Al reducir la varianza se busca que el modelo sea más estable ante cambios en los datos.

- En el método de Bagging los algoritmos son usados en **paralelo**.

- Este método utiliza el hecho de que hay cierta independencia entre cada unos de los algoritmos más simples, en los cuales se obtiene un predicción independiente y luego se computa, bien sea con la clase mayoritaria en problemas de clasificación o promediando en el caso de problemas de regresión.

- Es como si se usarán varias personas para tomar una decisión y se toma esta basada en lo que dice la mayoría.

**Pasos de implementación del bagging**

- Paso 1: Se crean múltiples subconjuntos a partir del conjunto de datos original con tuplas iguales, seleccionando observaciones con reemplazo.

- Paso 2: Se crea un modelo base en cada uno de estos subconjuntos.

- Paso 3: cada modelo se aprende en paralelo con cada conjunto de entrenamiento y de forma independiente entre sí.

- Paso 4: Las predicciones finales se determinan combinando las predicciones de todos los modelos.

### <FONT COLOR="blue"> 1.1.1 BaggingClassifier </FONT>

- Método basado generalmente en árboles de decisión. Pero puede utilizarse regresión logística, knn, etc.

- Toma muestras aleatorias con reemplazo (Bootstrapping)

- No necesariamente hay selección aleatoria de variables.






### <FONT COLOR="blue"> 1.1.2 Random Forest </FONT>

- Método basado en árboles de decisión

- Cada árbol se entrena con un bootstrap  (muestra aleatoria con reemplazo)

- En cada división del árbol, se selecciona un conjunto aleatorio de variables.

- Para hacer las divisiones en los nodos, se utiliza Gini o Entropia para hacer la mejor división posible.

- Tiene un menor sesgo

### <FONT COLOR="blue"> 1.1.3 Extra Tree (Extremely Randomized  Trees) </FONT>

- Método basado en árboles de decisión

- Por defecto **no usa bootstrasp**, es decir no toma una muestra aleatoria de datos con reemplazo, sino que usa todo el conjunto de datos. Pero se puede configurar para que haga bootstrap.

- Hace selección de variables por aleatorias por nodo.

- *Esto es lo más diferencial:* cuando se hacen árboles de decisión se usa indice de Gini o Entropía para hacer las mejores divisiones. En el caso de *Extra-Tree* los umbrales de corte con **aleatorios**.

- Tiene mayor sesgo pero menor varianza.

En conclusión

$$\text{Random Forest = BaggingClassifier con arboles de decision + seleccion aleatoria de variables en cada nodo}$$

$$\text{Extra Tree = Random Forest pero con divisiones de nodo aleatorias}$$

## <FONT COLOR="blue"> 1.2 Algoritmos de Boosting</FONT>

- El principio de este método es que los algoritmos operan de manera **secuencial**, es decir, cada modelo va operando y su predicción será tenida en cuenta para el siguiente modelo. Hay una dependencia entre modelos.

- Cuando se usa se manera secuencial, se tiene en cuenta los errores cometidos por el modelo anterior para hacer correcciones.

- Es decir, es como si para resolver un problema usamos el conocimiento de los errores de los otros para mejorar en algo lo que hacemos en la solución del problema.

- El objetivo principal de este algoritmo es reducción del **sesgo**. Es decir, se busca un modelo que se ajusta mejor a los datos. Se empieza con un modelo simple, se revisa donde se equivocó y se entrena un nuevo modelo que trata de ajustar esos errores. Luego, se repite esto muchas veces.

<center><img src="https://machinelearningparatodos.com/wp-content/uploads/2020/02/boosting.png" alt="centered image" width="300" height="300"></center>

- La diferencia con el bagging y el boosting es que este último no entrena los algorimtos independientemente, sino que se ponderan según los errores de los anteriores.

**Pasos en la implementación de Boosting**

Paso 1: Inicialice $M$, el número máximo de modelos que se ajustarán, y configure el contador de la iteración con $m=1$. Inicializar los pesos de observación $w_i=\frac{1}{N}$ para $i = 1, 2, ..., N$. Inicializace el modelo de  _ensemble_ con  $\hat{F_0}= 0$.

Paso 2: Usando pesos para las instancias $w_1,w_2,...,w_{N}$, entre un modelo $\hat{f}_m$ entrene un modelo que ***minimice el error ponderado*** $e_m$, el cual es definido sumando los pesos de las instancias mal clasificadas.

Paso 3: Adicione a el modelo de _ensemble:_

$$\hat{F}_m=\hat{F}_{m-1}+\alpha_{m}\hat{f}_{m} \ \ \ \  \text{ donde $α_m=\frac{log(1-e_m)}{e_m}$}$$

Paso 4: Actualizar los pesos $w_1,w_2,..., w_N$ de modo que se aumentan los pesos para las instancias que fueron mal clasificadas. El tamaño del incremento depende de $α_m$, con valores más grandes de $α_m$ conducen a pesos más grandes.

Paso 5: Incrementar el contador del modelo $m=m+1$. Si $m≤M$ vaya al paso 2.

La estimación _Boosting_ viene dada por:

$$\hat{F} = α_1\hat{f}_1+ α_2\hat{f}_2 + ⋯ + α_M\hat{f}_{M}$$

Al aumentar los pesos de las observaciones que fueron mal clasificadas, el algoritmo obliga a los modelos a entrenar más intensamente en los datos para los que se desempeñó de manera deficiente. El factor $α_m$ asegura que los modelos con menor error tengan mayor peso. El aumento de gradiente es similar a Adaboost pero presenta el problema como una optimización de una función de coste. En lugar de ajustar los pesos, el _gradient Boosting_ ajusta los modelos a un _pseudo-residual_, que tiene el efecto de entrenar más intensamente en los residuos más grandes. En el espíritu del _Random Forrest_, _stocastic gradient boosting_ agrega aleatoriedad a el algoritmo mediante el muestreo de instancias y variables predictoras en cada etapa.

<FONT COLOR="magenta"> Algoritmos basados en Boosting </FONT>

Algunos de los algoritmos clásicos basados en Boosting son:

- **AdaBoost (Adaptive Boosting):** ajusta pesos de las muestras en cada iteración, con el propósito de aumentar la probabilidad de que un dato sea usado al entrenar el siguiente árbol. Pone más peso en errores.

- **Gradient Boosting Machines (GBM):** entrena modelos para minimizar una función de pérdida mediante gradiente descendente. Usa gradiente de pérdida.

- **XGBoost (Extreme Gradient Boosting):** versión optimizada y regularizada de GBM. GBM con reguralización. Es muy popular.

- **LightGBM:** GBM rápido y eficiente, muy usado en conjuntos de datos grandes.

- **CatBoost**  GBM optimizado para manejar variables categóricas directamente.

**Recuerde que:**

- En problemas de **clasificación** los errores se calculan por medio de la función de entropía cruzada o Log loss

$$LogLoss = -\dfrac{1}{n} \sum \limits_{i=1}^n[y_i \log(p_i)+(1-y_i)\log(1-p_i)]$$

- En problemas de **regresión** los errores están dados por el error cuadrático medio

$$MSE = \dfrac{1}{n} \sum \limits_{i=1}^{n}(y_i-\hat{y}_i)^2$$
