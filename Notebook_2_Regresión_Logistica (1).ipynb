# <FONT SIZE=5 COLOR="purple"> 1. Explicación del Modelo </FONT>

- Usamos regresión logística para estudiar la probabilidad de una variable categórica cualitativa (por ejemplo, dicotómica)  en función de una o más variables cuantitativas.

- En otras palabras, la regresión logística no permite clasificar directamente, pero si permite determinar una probabilidad de pertenecer a un grupo determinado de la variable categórica. Y luego, de acuerdo con esa probabilidad, tomar una decisión de asignación.

Se puede emplear la regresión logística por ejemplo:

- La probabilidad de estar empleado en función de carrera, nivel de inglés, edad.

- La probabilidad de que alguien compre un producto en función del ingreso, estrato, género.

- La probabilidad de pagar una deuda en funcion de otras deudas, ingreso, estado civil.

- La probabilidad de que un paciente tenga o no diabetes

Supongamos que tenemos una variable categórica con dos clases, que las llamaremos $\{0, 1\}$. Entonces, si tratamos de buscar un modelo lineal de la forma

$$y = \beta_0 +\beta_1 x_1 +\epsilon$$

donde $y$ es una variable binaria, no sería un buen modelo ya que la parte izquierda de la expresión es continua. Luego, podríamos pensar en no tener una relación directa, sino vía la probabilidad, así:

$$P(y=1 \mid X = x_i)= \beta_0+\beta_1x_i$$

Cuando intentamos usar mínimos cuadrados, es fácil ver que el modelo no funciona ya que predecimos valores mayores que 1 y menores que 0.

<center><img src="https://github.com/Fabian830348/cursos/blob/0da91b828112d77dbddb387cbb7043ab8a78f252/Imagen/log1.png?raw=true" alt="centered image" width="500" height="400"></center>

podemos observar que este modelo no se ajusta a los datos. Más aún, si tratamos de predecir un valor, como $x=1.1$, al tratarse de una probabilidad, este valor no tiene sentido, pues es negativo.

Entonces, buscaremos una función cumpla lo siguiente

- $P(y=1 \mid X = x_i) = \sigma(\beta_0 + \beta_1 x_i)$

- $P(y=1 \mid  x_i)$ sea creciente en $x$ para $\beta_1>0$.

- $P(y=1 \mid  x_i)$ sea decreciente en $x$ para $\beta_1<0$.

- $0 \leq P(y=1 \mid  x_i) \leq 1$ para todo valor de $x_i$.

Se puede ver rápidamente que la función debería tener una forma de S. A esta función se le denomina función sigmoide y es la base del modelo logístico

$$\sigma(x)= \dfrac{1}{1+e^{-x}}=\dfrac{e^x}{1+e^x}$$

La gráfica de esta función es:

<center><img src="https://github.com/Fabian830348/cursos/blob/0da91b828112d77dbddb387cbb7043ab8a78f252/Imagen/log2.png?raw=true" alt="centered image" width="500" height="400"></center>

Así pues, tenemos una expresión de la forma

$$\hat{y}=P(y=1 \mid  X = x_i)= \dfrac{e^{\beta_0+\beta_1x_i}}{1+e^{\beta_0+\beta_1x_i}} = \sigma (\beta_0+\beta_1x_i)$$

Ahora bien, recordemos que queremos que los valores de salida sean muy cercanos a los valores reales. Pero en este caso, ¿qué significa esto?

- En problemas de clasificación binaria, el valor real $y_i \in \{0,1\}$. Ya que toma dos valores posibles.

- La probabilidad estimada por el modelo será $\hat{y_i}=P(y=1 \mid  X = x_i)$

Así que tenemos dos casos

1. Caso 1: $y_i = 1$

 La probabilidad de observar ese dato es simplemente la probabilidad de la clase 1

 $$P(y_i=1 \mid  X = x_i) = \hat{y_i}$$

2. Caso 1: $y_i = 0$

 La probabilidad de observar ese dato es simplemente la probabilidad de la clase 1

 $$P(y_i=0 \mid  X = x_i) = 1-\hat{y_i}$$

Para no escribir dos fórmulas, lo que hacemos es juntarlas mediante la expresión.

$$P(y_i \mid  X = x_i) = \hat{y_i}^{y_i}(1- \hat{y_i})^{1-y_i} $$

Lo anterior es la probabilidad de observar el dato $y_i$, pero como son n-datos, se define la **verosimilitud del conjunto de datos como**

$$L(\beta_0, \beta_1) = \prod_{i=1}^{n} \hat{y_i}^{y_i}(1- \hat{y_i})^{1-y_i} $$

Ahora, si aplicamos logaritmo obtenemos **log_verosimilitud**

$$\ell (\beta_0, \beta_1) = \log (L(\beta_0,\beta_1)) = \sum \limits_{i=1}^n [y_i\log(\hat{y_i})+(1-y_i)\log(1-\hat{y_i})]$$

Lo que busca la regresión logística es **máximizar** las probabilidades del conjunto de datos. En otras palabras ajustar los valores de $\beta_0$ y $\beta_1$ para que las probabilidades asignadas a las clases correctas de cada ejemplo sea lo más alta posible.

En machine learning, en general no se busca maximizar si no minimizar algo. En este caso, se quiere minimizar la **entropia cruzada**, que es:

$$Entropia\_Cruzada = -\dfrac{1}{n} \ell(\beta_0, \beta_1) = -\dfrac{1}{n}\sum \limits_{i=1}^n [y_i\log(\hat{y_i})+(1-y_i)\log(1-\hat{y_i})] $$

De está manera, en la regresión logística (y en general) para problemas de clasificación, el objetivo es minimizar la función de entropia cruzada. A esta función se le denomina **función de pérdida**.

<FONT SIZE=5 COLOR="blue"> LOG de Odds </FONT>

Si $P(y_i=y_0 \mid   X = x_i)$ es la probabilidad de que la variable cualitativa $y$ adquiera el valor $y_0$ (sin perdida de generalidad tomamos uno, $y_0=1$) dado que el predictor $X$ toma el valor de $x_i$

Vamos a llamar $U=P(y_i=1 \mid  x_i)$, entonces

$$ \begin{align*} U & = \dfrac{e^{\beta_0+\beta_1x_i}}{1+e^{\beta_0+\beta_1x_i}} \\ ({1+e^{\beta_0+\beta_1x_i}} )U & = e^{\beta_0+\beta_1x_i} \\U +e^{\beta_0+\beta_1x_i}U & = e^{\beta_0+\beta_1x_i} \\ e^{\beta_0+\beta_1x_i}U-e^{\beta_0+\beta_1x_i} & = -U \\ e^{\beta_0+\beta_1x_i}(U-1)& = -U \\ e^{\beta_0+\beta_1x_i} & = \dfrac{-U}{U-1} \\ \beta_0+\beta_1x_i & = \ln \left ( \dfrac{U}{1-U}  \right ) \end{align*}    $$


$$\beta_0+\beta_1 x_i= \ln \left ( \dfrac{P(y_i=1 \mid  x_i)}{1-P(y_i=1 \mid  x_i)}  \right )$$

A esta última expresión se le conoce como el $LOG$ de $Odds$.

Para interpretar este logaritmo tengamos en cuenta lo siguiente. Si la probabilidad de que suceda un evento es $0.7$, entonces la probabilidad de que no suceda el evento es $1-0.7$, es decir $0.3$. Los Odds o razón de probabilidad de que se cumpla el evento, es el cociente entre la probabilidad de que sea verdadero entre la que sea falso.

## <FONT SIZE=5 COLOR="purple"> 2.1 Hiperparámetros de la regresión logística </FONT>

La regresión logística tiene los siguientes hiperparámetros

1. *penalty* : Parámetro de regularización. Es una técnica para evitar el sobreajuste (**overfitting)**

$$Funcion\_de\_Costo = Log\_Loss + Penalizacion$$

Esta penalización puede ser $L1$ que se llama Lasso y $L2$ que se llama Ridge

**Regularización *L2* (Ridge)**

Penaliza la suma de los cuadrados de los coeficientes

$$\lambda \sum \limits_{j=1}^n \beta_{j}^2$$

   - Empuja los coeficientes hacia valores pequeños, pero no los anula

   - Mantiene todas las variables del modelo.

**Regularización *L1* Lasso**

Penaliza la suma del valor absoluto de los coeficientes

$$\lambda \sum \limits_{j=1}^n |\beta_{j}|$$

   - Tiene el efecto de que algunos coeficientes se vuelven cero.

   - Selección automática de variables.

**Elastic Net : *L1+L2***

$$\lambda_1 \sum \limits_{j=1}^n \beta_{j}^2+\lambda_2 \sum \limits_{j=1}^n |\beta_{j}|$$


En resumen, la función de pérdida queda

$$ -\dfrac{1}{n}\sum \limits_{i=1}^n [y_i\log(\hat{y_i})+(1-y_i)\log(1-\hat{y_i})] + \lambda \sum \limits_{j=1}^n \beta_{j}^2$$

o

$$ -\dfrac{1}{n}\sum \limits_{i=1}^n [y_i\log(\hat{y_i})+(1-y_i)\log(1-\hat{y_i})] + \lambda \sum \limits_{j=1}^n |\beta_{j}|$$

2. *Solver*: Parámetro de optimización. Algoritmo utilizado para optimizar

 - *liblinear* : soporta L1 y L2. Para conjunto pequeños de datos.

 - *saga*: conjuntos de datos grandes. Soporta L1, L2 y Elasticnet.

 - *lbfga*: problemas multiclase con L2

 - *newtin-cg*. Es útil en multiclase.

3. Hiperparámetro *C*

En *LogisticRegression* de scikit-learn, no se usa $\lambda$ sino el inverso

$$C = \dfrac{1}{\lambda}$$

  - C grande, implica $\lambda$ pequeño: menos regularización.

  - C pequeño, implica $\lambda$ grande: más regularización.

  - C = [0.001, 0.01 , 0.1 , 1 , 10 , 100 , 1000]

Podemos usar algunas librerías de Python que nos permiten hacer el proceso más eficiente.

````python
variables_num = ["Edad", "Salario"]
variables_cat = ["Sexo"]
````

````python
from sklearn.preprocessing      import OneHotEncoder
from sklearn.compose            import ColumnTransformer

preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), variables_num),
    ('cat', OneHotEncoder(), variables_cat) #passthrough: si no queremos que haga nada con algunas columnas #drop = "first"
])
````

````python
# hacer el dataframe
pd.DataFrame(preprocessor.fit_transform(df),
             columns=preprocessor.get_feature_names_out())
````

````python
# 1. Definir las variables
# las variables predictoras
X = df.drop(columns=["Compra"], axis = 1)
# la variable objetivo
y = df["Compra"]
# 2. Conjunto de entrenamiento y de prueba
X_train, X_test, y_train, y_test = train_test_split(X,                        # variables predictoras
                                                    y,                        # variable de respuesta
                                                    stratify=y,               # estratificamos con respecto a y (asegura que se mantenga la proporción de clases en y)
                                                    random_state = 42,         # semilla para que al ejecutar siempre de igual
                                                    test_size = 0.3)          # tamaño del conjunto de prueba

# 3. Definimos el preprocesador de escalamiento y codificación de variables categóricas
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), variables_num),
    ('cat', OneHotEncoder(drop = "first"), variables_cat) #passthrough
])
````

````python
# 6. Calcular las predicciones
y_pred_log = modelo_log.predict(X_test)

# 7. matrix de confusión
print("Matriz de Confusión \n\n", confusion_matrix(y_test, y_pred_log), "\n")

# 8. métricas
metrics=["accuracy", "recall" , "precision", "f1"]
values = [accuracy_score(y_test,y_pred_log),
          recall_score(y_test,y_pred_log),
          precision_score(y_test,y_pred_log),
          f1_score(y_test,y_pred_log)]
print("Métricas \n", pd.DataFrame({"metrics": metrics , "values" : values}), "\n")

# 9. Resumen de las métricas
print("Resumen evaluación \n", classification_report(y_test, y_pred_log))
````

Recordemos

1. Los **Odds** o razones de probabilidad

$$Odds = \dfrac{P(evento)}{1-P(evento)}$$

Si *evento* = "hacer una compra" y la probabilidad de hacer una compra es $0.7$ $(70\%)$ los odds son:

$$Odds = \dfrac{0.7}{1-0.7} = \dfrac{0.7}{0.3} = 2.33$$

es decir, la probabilidad de hacer una compra es 2.33 veces mayores que las probabilidades de que no ocurra.

2. Los Odds Ratio (razón de odds) indica cómo cambia la probabilidad de que ocurra un evento cuando la variable aumenta en una unidad.

$$Odds\_Ratio = \dfrac{Odds\_antes}{Odds\_despues}$$

En regresión logística es:

$$Odds\_Ratio = e^{\beta_i}$$

Interpretación del Odds Ratio:

- $Odds\_Ratio = 1$ : No hay diferencia en las odds entre los dos grupos.

- $Odds\_Ratio > 1$: El evento es más probable en el grupo 1 en comparación con el grupo 2.

- $Odds Ratio < 1$: El evento es menos probable en el grupo 1 en comparación con el grupo 2.

3. Log(odds) nos indica cómo cambia el logaritmo de las probabilidades de que ocurra un evento (en este caso, realizar una compra) cuando una variable aumenta en una unidad, manteniendo las demás constantes.

$$Log(Odds)$$

De está manera tenemos

1. **Sexo**

- Supongamos que estamos analizando el efecto del sexo en la probabilidad de realizar una compra, donde Mujer = 1 y hombre = 0. El coeficiente para la variable Sexo_mujer en la regresión logística es $\beta = -0.42$ y el

 $$Odds\_Ratio = e^{-0.42} = 0.65 < 1 $$

Esto significa que ser mujer reduce la probabilidad de compra.

2. **Edad**

- Coeficiente (Log-odds) = 2.20 y $Odds\_Ratio = e^{2.20} = 9.04 > 1 $.

Seleccionamos los conjuntos de entrenamiento y prueba

````python
# Dividir en dos conjuntos
# las variables predictoras
X = cancer.iloc[:, 1:]
# la variable objetivo
y = cancer.iloc[:, 0]
X_train, X_test, y_train, y_test = train_test_split(X,                        # variables predictoras
                                                    y,                        # variable de respuesta
                                                    random_state = 0,         # semilla para que al ejecutar siempre de igual
                                                    test_size = 0.2,          # tamaño del conjunto de prueba
                                                    stratify = y,             # estratificar por la variable de respuesta
                                                    shuffle= True)            # barajear
````

````python
X_train
````

Vamos a escalar los datos. Aplicaremos el escalador *max-min*,

$$\dfrac{X- min(X)}{max(X)-min(X)}$$

Se puede usar otro escalador como el *StandarScaler*

````python
escalar = MinMaxScaler()
X_train_s = escalar.fit_transform(X_train)
X_test_s = escalar.transform(X_test)
````

Ahora, generamos el modelo

````python
from sklearn.linear_model import LogisticRegression
Log_reg = LogisticRegression(random_state=0)
Log_reg.fit(X_train_s, y_train)
````

Veamos algunos coeficientes del modelo

````python
# coeficientes de los betas
print(Log_reg.coef_)
# coeficiente del beta_0
print(Log_reg.intercept_)
````

Revisaremos como se comporta el modelo en el conjunto de entrenamiento y de prueba



````python
# revisamos las métricas en los conjuntos de entrenamiento
y_pred_train=Log_reg.predict(X_train_s)
print("En los datos de entrenamiento \n \n",classification_report(y_train, y_pred_train))

# revisamos las métricas en el conjunto de prueba
# tenga presente que este funciona con un umbral del 50%
y_pred_test=Log_reg.predict(X_test_s)
print("En los datos de prueba \n \n",classification_report(y_test, y_pred_test))
````

Podemos hacer la matriz de confusión en ambos conjuntos



````python
# matrices de confusión
cm1 = confusion_matrix(y_train, y_pred_train)
cm2 = confusion_matrix(y_test, y_pred_test)

# Crear una figura con dos subgráficos
fig, axes = plt.subplots(1, 2, figsize=(7, 3))

# Graficar la primera matriz de confusión
sns.heatmap(cm1,
            annot=True,
            annot_kws = {'size':10},
            cmap='Blues',
            fmt="g",
            ax=axes[0])
axes[0].set_title('Matriz de Confusión Train Log_reg')
axes[0].set_xlabel('Predicción')
axes[0].set_ylabel('Real')

# Graficar la segunda matriz de confusión
sns.heatmap(cm2,
            annot=True,
            annot_kws = {'size':10},
            cmap='Blues',
            fmt="g",
            ax=axes[1])
axes[1].set_title('Matriz de Confusión Test Log_reg')
axes[1].set_xlabel('Predicción')
axes[1].set_ylabel('Real')

# Mostrar el gráfico
plt.tight_layout()
plt.show()
````
